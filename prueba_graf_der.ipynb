```python

import os
import re
import math
import time
import logging
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt



# User files location


FOLDER = r"C:\Users\raulm\Downloads"

TARGET_SCHEME = "Blackadar"

DO_CALIBRATION = True

LOG_DEBUG = True

RANDOM_SEED = 42

PLOT_PHYSICAL_DIAGNOSTICS = False

# Additional analysis diagnostics 
PLOT_RATE_DIAGNOSTICS = True
PLOT_BRUNT_VAISALA = True
PRINT_ERROR_METRICS = True
FINAL_SLOPE_FRACTION = 0.12



# Calibration given no uniform data length


N_RANDOM_STAGE1 = 70
N_REFINE_STAGE1 = 20

N_RANDOM_STAGE2 = 90
N_REFINE_STAGE2 = 25


N_RANDOM_STAGE3 = 120
N_REFINE_STAGE3 = 32


WEIGHTS_GLOBAL = (
    0.30,   
    0.45,   
    0.25,   
)



HYBRID_GUIDANCE = True

GUIDANCE_ALPHA = 0.12

GUIDANCE_CLIP = 0.35



GUIDANCE_STAGE_STRENGTHS = {

    "stage1": {
        "liq": 0.055,
        "pbl": 0.015,
        "ri": 0.015,
    },

    "stage2": {
        "liq": 0.020,
        "pbl": 0.040,
        "ri": 0.018,
    },

    
    "stage3": {
        "liq": 0.000,
        "pbl": 0.000,
        "ri": 0.000,
    },
}



SEG_WEIGHTS_WATER = (
    (0.00, 0.18, 1.00),
    (0.18, 0.62, 1.10),
    (0.62, 1.00, 1.15),
)

SEG_WEIGHTS_PBL = (
    (0.00, 0.22, 1.15),
    (0.22, 0.65, 1.00),
    (0.65, 0.88, 1.35),
    (0.88, 1.00, 1.15),
)

SEG_WEIGHTS_RI = (
    (0.00, 0.30, 1.10),
    (0.30, 0.72, 1.10),
    (0.72, 1.00, 1.10),
)



random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s %(levelname)s %(message)s"
)

log = logging.getLogger(
    "fog_blackadar_hybrid_v5"
)

if LOG_DEBUG:
    log.setLevel(logging.DEBUG)



# CSV file reading
# NO MODIFICAR


SEPARATORS = [";", ",", "\t", "|"]

num_token_re = re.compile(
    r'[+\-]?[0-9\.,eE\+\-()]+'
)


def normalize_text(s: str) -> str:

    if s is None:
        return ""

    s = str(s)

    s = (
        s
        .replace('\u2212', '-')
        .replace('\u2012', '-')
        .replace('\u2013', '-')
    )

    s = (
        s
        .replace('\u2014', '-')
        .replace('\u2010', '-')
    )

    s = (
        s
        .replace('\u00A0', ' ')
        .replace('\u202f', ' ')
    )

    return s.strip()


def parse_to_float(s: str):

    if s is None:
        return None

    s0 = normalize_text(s)

    if s0 == "":
        return None

    negative = False

    if (
        s0.startswith('(')
        and s0.endswith(')')
    ):

        negative = True

        s0 = (
            s0[1:-1]
            .strip()
        )

    if (
        (
            s0.startswith('"')
            and s0.endswith('"')
        )
        or
        (
            s0.startswith("'")
            and s0.endswith("'")
        )
    ):

        s0 = (
            s0[1:-1]
            .strip()
        )

    sign = ''

    if (
        s0.startswith('+')
        or s0.startswith('-')
    ):

        sign = s0[0]

        s0 = (
            s0[1:]
            .strip()
        )

    if (
        '.' in s0
        and ',' in s0
    ):

        if (
            s0.rfind(',')
            > s0.rfind('.')
        ):

            s_clean = (
                s0
                .replace('.', '')
                .replace(',', '.')
            )

        else:

            s_clean = (
                s0
                .replace(',', '')
            )

    elif (
        ',' in s0
        and '.' not in s0
    ):

        s_clean = (
            s0
            .replace('.', '')
            .replace(',', '.')
        )

    else:

        s_clean = s0

    if sign:
        s_clean = sign + s_clean

    if negative:

        s_clean = (
            '-'
            + s_clean
            if not s_clean.startswith('-')
            else s_clean
        )

    s_fin = re.sub(
        r'[^0-9\.\-+eE]',
        '',
        s_clean
    )

    if s_fin in [
        '',
        '+',
        '-'
    ]:

        return None

    try:

        val = float(s_fin)

        return (
            val
            if math.isfinite(val)
            else None
        )

    except:

        return None


def detect_sep_decimal(
    text,
    n_lines=10
):

    sample = "\n".join(
        text.splitlines()[:n_lines]
    )

    counts = {
        s: sample.count(s)
        for s in SEPARATORS
    }

    decimal_comma_patterns = len(
        re.findall(
            r'\d+,\d+',
            sample
        )
    )

    if (
        counts.get(';', 0) > 0
        and decimal_comma_patterns > 0
    ):

        return ';', ','

    if (
        counts.get('\t', 0) > 0
        and decimal_comma_patterns > 0
    ):

        return '\t', ','

    best = max(
        counts,
        key=lambda k: counts[k]
    )

    decimal = (
        ','
        if (
            decimal_comma_patterns > 0
            and best != ','
        )
        else '.'
    )

    return best, decimal


def try_read_with(
    path,
    encoding,
    sep,
    decimal
):

    try:

        df = pd.read_csv(
            path,
            sep=sep,
            header=None,
            engine='python',
            decimal=decimal,
            encoding=encoding,
            skip_blank_lines=True,
            skipinitialspace=True,
            low_memory=False
        )

        for c in df.select_dtypes(
            include=['object']
        ).columns:

            df[c] = (
                df[c]
                .astype(str)
                .map(normalize_text)
            )

        return df

    except Exception:

        return None


def read_force_second_column_v3(
    path,
    debug=False
):

    path = str(path)

    raw = None

    used_enc = None

    encodings = [
        'utf-8',
        'latin1',
        'cp1252',
        'iso-8859-1'
    ]

    for enc in encodings:

        try:

            with open(
                path,
                'r',
                encoding=enc,
                errors='replace'
            ) as f:

                raw = f.read()

            used_enc = enc

            break

        except:

            raw = None

    if raw is None:

        raise ValueError(
            "No pude leer archivo "
            "(encs probadas) "
            + path
        )

    guessed_sep, guessed_dec = (
        detect_sep_decimal(raw)
    )

    sep_order = (
        [guessed_sep]
        + [
            s
            for s in SEPARATORS
            if s != guessed_sep
        ]
    )

    df_good = None

    used_sep = None

    used_dec = None

    for enc in encodings:

        for sep in sep_order:

            for dec in [
                guessed_dec,
                (
                    ','
                    if guessed_dec == '.'
                    else '.'
                )
            ]:

                df_try = try_read_with(
                    path,
                    enc,
                    sep,
                    dec
                )

                if df_try is None:
                    continue

                if df_try.shape[1] == 1:

                    col0 = (
                        df_try.iloc[:, 0]
                        .astype(str)
                    )

                    if (
                        col0
                        .str
                        .contains(';')
                        .any()
                    ):

                        df_try2 = (
                            col0
                            .str
                            .split(
                                ';',
                                expand=True
                            )
                        )

                        if df_try2.shape[1] >= 2:
                            df_try = df_try2

                    elif (
                        col0
                        .str
                        .contains(r'\s+')
                        .any()
                    ):

                        df_try2 = (
                            col0
                            .str
                            .split(
                                r'\s+',
                                expand=True
                            )
                        )

                        if df_try2.shape[1] >= 2:
                            df_try = df_try2

                if df_try.shape[1] < 2:
                    continue

                arr0 = (
                    df_try.iloc[:, 0]
                    .astype(str)
                    .tolist()
                )

                arr1 = (
                    df_try.iloc[:, 1]
                    .astype(str)
                    .tolist()
                )

                start_idx = 0

                found = False

                for si in range(
                    min(
                        10,
                        len(arr0)
                    )
                ):

                    a = parse_to_float(
                        arr0[si]
                    )

                    b = parse_to_float(
                        arr1[si]
                    )

                    if (
                        a is not None
                        and b is not None
                    ):

                        start_idx = si
                        found = True
                        break

                if not found:

                    for si in range(
                        min(
                            30,
                            len(arr0)
                        )
                    ):

                        a = parse_to_float(
                            arr0[si]
                        )

                        b = parse_to_float(
                            arr1[si]
                        )

                        if (
                            a is not None
                            and b is not None
                        ):

                            start_idx = si
                            found = True
                            break

                xs = []
                ys = []

                for r in range(
                    start_idx,
                    len(arr0)
                ):

                    a = parse_to_float(
                        arr0[r]
                    )

                    b = parse_to_float(
                        arr1[r]
                    )

                    if (
                        a is None
                        or b is None
                    ):
                        continue

                    xs.append(a)
                    ys.append(b)

                if len(xs) == 0:

                    for r in range(
                        start_idx,
                        len(arr0)
                    ):

                        a = parse_to_float(
                            arr1[r]
                        )

                        b = parse_to_float(
                            arr0[r]
                        )

                        if (
                            a is None
                            or b is None
                        ):
                            continue

                        xs.append(a)
                        ys.append(b)

                if len(xs) == 0:
                    continue

                med_y = (
                    np.median(
                        np.abs(ys)
                    )
                    if len(ys) > 0
                    else 0.0
                )

                max_y = (
                    np.max(
                        np.abs(ys)
                    )
                    if len(ys) > 0
                    else 0.0
                )

                ratio = (
                    max_y
                    / (
                        med_y
                        if med_y > 0
                        else 1.0
                    )
                )

                if (
                    max_y > 1e12
                    or ratio > 1e9
                ):
                    continue

                df_good = (
                    np.array(
                        xs,
                        dtype=float
                    ),
                    np.array(
                        ys,
                        dtype=float
                    )
                )

                used_sep = sep
                used_dec = dec
                used_enc = enc

                break

            if df_good is not None:
                break

        if df_good is not None:
            break

    if df_good is None:

        xs = []
        ys = []

        for L in raw.splitlines():

            sline = normalize_text(L)

            toks = (
                num_token_re
                .findall(sline)
            )

            if len(toks) >= 2:

                a = parse_to_float(
                    toks[0]
                )

                b = parse_to_float(
                    toks[1]
                )

                if (
                    a is not None
                    and b is not None
                ):

                    xs.append(a)
                    ys.append(b)

                    continue

            for sep in [
                ';',
                ',',
                '\t',
                '|',
                ' '
            ]:

                parts = [
                    p.strip()
                    for p in sline.split(sep)
                    if p.strip() != ''
                ]

                if len(parts) >= 2:

                    a = parse_to_float(
                        parts[0]
                    )

                    b = parse_to_float(
                        parts[1]
                    )

                    if (
                        a is not None
                        and b is not None
                    ):

                        xs.append(a)
                        ys.append(b)
                        break

        if len(xs) == 0:

            raise ValueError(
                "No pude extraer pares X-Y de: "
                + path
            )

        df_good = (
            np.array(
                xs,
                dtype=float
            ),
            np.array(
                ys,
                dtype=float
            )
        )

        used_sep = 'fallback'
        used_dec = 'fallback'

    if LOG_DEBUG:

        log.debug(
            f"[read_v3] "
            f"{os.path.basename(path)} "
            f"-> pares={len(df_good[0])} "
            f"(enc={used_enc} "
            f"sep={used_sep} "
            f"dec={used_dec})"
        )

    return (
        df_good[0],
        df_good[1]
    )



# Physics constants


g = 9.81

cp = 1004.0

Lv = 2.5e6

rho0 = 1.2

epsilon = 0.97

sigma = 5.67e-8

S0 = 600.0

nz = 100

dz = 1.0

z = np.linspace(
    0.0,
    dz * (nz - 1),
    nz
)

lat_deg = 40.0

Omega = 7.292115e-5

f = (
    2.0
    * Omega
    * math.sin(
        math.radians(
            lat_deg
        )
    )
)

ug_g = 5.0
vg_g = 0.0



# Basic physics


def sat_mix(T, p=1e5):

    Tc = (
        np.asarray(
            T,
            dtype=float
        )
        - 273.15
    )

    es = (
        610.0
        * np.exp(
            17.27
            * Tc
            / (
                Tc
                + 237.3
            )
        )
    )

    return (
        0.622
        * es
        / p
    )


def running_mean(a, window=5):

    a = np.asarray(
        a,
        dtype=float
    )

    if (
        a.size < window
        or window <= 1
    ):
        return a.copy()

    kernel = (
        np.ones(
            window
        )
        / window
    )

    return np.convolve(
        a,
        kernel,
        mode="same"
    )


def resample_series(a, n):

    a = np.asarray(
        a,
        dtype=float
    )

    if n <= 0:
        return np.array([])

    if a.size == 0:
        return np.zeros(n)

    if a.size == 1:
        return np.full(
            n,
            a[0]
        )

    xo = np.linspace(
        0.0,
        1.0,
        len(a)
    )

    xn = np.linspace(
        0.0,
        1.0,
        n
    )

    return np.interp(
        xn,
        xo,
        a
    )


def resample_to_len(a, n):

    return resample_series(
        a,
        n
    )


def norm01(x):

    x = np.asarray(
        x,
        dtype=float
    )

    x = np.nan_to_num(
        x,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    if x.size == 0:
        return x

    x = (
        x
        - np.min(x)
    )

    xmax = np.max(x)

    if xmax <= 1e-12:
        return np.zeros_like(x)

    return (
        x / xmax
    )


def safe_mean(x, default=0.0):

    x = np.asarray(
        x,
        dtype=float
    )

    if x.size == 0:
        return default

    value = np.nanmean(x)

    if not np.isfinite(value):
        return default

    return float(value)


def safe_max(x, default=0.0):

    x = np.asarray(
        x,
        dtype=float
    )

    if x.size == 0:
        return default

    value = np.nanmax(x)

    if not np.isfinite(value):
        return default

    return float(value)


def guidance_scale(
    x,
    fallback=1.0
):

    x = np.asarray(
        x,
        dtype=float
    )

    if len(x) < 2:
        return max(
            fallback,
            1e-6
        )

    scale = float(
        np.std(x)
    )

    if (
        not np.isfinite(scale)
        or scale < 1e-6
    ):

        scale = float(
            np.max(x)
            - np.min(x)
        )

    if (
        not np.isfinite(scale)
        or scale < 1e-6
    ):
        scale = fallback

    return max(
        scale,
        1e-6
    )



# Diffusion


def diffusive_tendency(
    field,
    Km
):

    phi = np.asarray(
        field,
        dtype=float
    )

    K = np.asarray(
        Km,
        dtype=float
    )

    tendency = np.zeros_like(
        phi
    )

    if len(phi) < 3:
        return tendency

    Kface = (
        0.5
        * (
            K[:-1]
            + K[1:]
        )
    )

    flux = np.zeros(
        len(phi) + 1
    )

    for i in range(
        1,
        len(phi)
    ):

        flux[i] = (
            -Kface[i - 1]
            * (
                phi[i]
                - phi[i - 1]
            )
            / dz
        )

    for i in range(
        len(phi)
    ):

        tendency[i] = (
            -(
                flux[i + 1]
                - flux[i]
            )
            / dz
        )

    return tendency



# Vertical profiles


def normalized_profile(depth):

    profile = np.exp(
        -z
        / max(
            depth,
            1e-3
        )
    )

    profile /= (
        np.trapezoid(
            profile,
            z
        )
        + 1e-12
    )

    return profile


def radiative_cooling_profile(
    Rnet,
    depth
):

    return (
        Rnet
        * normalized_profile(
            depth
        )
    )



# Wind


def init_wind_profile(
    perturb=0.0,
    z0=0.1
):

    ze = np.maximum(
        z,
        1e-3
    )

    denominator = np.log(
        1.0
        + z[-1]
        / z0
    )

    u = (
        ug_g
        * np.log(
            1.0
            + ze
            / z0
        )
        / denominator
    )

    v = np.full(
        nz,
        vg_g
    )

    u[0] = 0.0

    u += (
        perturb
        * ug_g
        * z
        / max(
            z[-1],
            1.0
        )
    )

    return (
        u,
        v
    )



# Richardson


def local_ri_profile(
    theta,
    u,
    v
):

    dtheta = np.gradient(
        theta,
        dz
    )

    du = np.gradient(
        u,
        dz
    )

    dv = np.gradient(
        v,
        dz
    )

    shear2 = (
        du**2
        + dv**2
        + 1e-10
    )

    Ri = (
        g
        / np.maximum(
            theta,
            1.0
        )
        * dtheta
        / shear2
    )

    return np.clip(
        Ri,
        -60.0,
        60.0
    )


def compute_ri_index(
    theta,
    u,
    v,
    params
):

    ri_ref = float(
        params.get(
            "ri_ref",
            20.0
        )
    )

    ri_scale = float(
        params.get(
            "ri_scale",
            12.0
        )
    )

    idx = int(
        np.clip(
            round(
                ri_ref / dz
            ),
            2,
            nz - 1
        )
    )

    theta0 = np.mean(
        theta[:2]
    )

    theta1 = np.mean(
        theta[
            idx - 1:
            idx + 1
        ]
    )

    u0 = np.mean(
        u[:2]
    )

    u1 = np.mean(
        u[
            idx - 1:
            idx + 1
        ]
    )

    v0 = np.mean(
        v[:2]
    )

    v1 = np.mean(
        v[
            idx - 1:
            idx + 1
        ]
    )

    shear2 = (
        (
            u1 - u0
        )**2
        + (
            v1 - v0
        )**2
        + 1e-10
    )

    Ri_bulk = (
        g
        / max(
            theta0,
            1.0
        )
        * (
            theta1 - theta0
        )
        * z[idx]
        / shear2
    )

    Ri_profile = local_ri_profile(
        theta,
        u,
        v
    )

    Ri_gradient = safe_mean(
        Ri_profile[
            1:min(
                8,
                nz
            )
        ],
        0.0
    )

    raw = (
        0.68
        * Ri_gradient
        + 0.32
        * Ri_bulk
    )

    raw = np.clip(
        raw,
        -2.8,
        2.8
    )

    return float(
        np.clip(
            ri_scale
            * raw,
            -35.0,
            35.0
        )
    )



# Comparison target


def transform_forcing_series(series):

    s = np.asarray(
        series,
        dtype=float
    )

    s = np.nan_to_num(
        s,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    s = np.maximum(
        s,
        0.0
    )

    short = running_mean(
        s,
        3
    )

    long_window = (
        11
        if len(s) >= 11
        else max(
            3,
            len(s) // 2 + 1
        )
    )

    long = running_mean(
        s,
        long_window
    )

    anomaly = (
        short
        - long
    )

    d1 = np.gradient(
        short
    )

    d2 = np.gradient(
        d1
    )

    jump = np.diff(
        short,
        prepend=short[0]
    )

    driver = np.clip(

        0.24
        * norm01(
            np.maximum(
                anomaly,
                0.0
            )
        )

        + 0.23
        * norm01(
            np.maximum(
                d1,
                0.0
            )
        )

        + 0.16
        * norm01(
            np.abs(
                d2
            )
        )

        + 0.19
        * norm01(
            np.maximum(
                jump,
                0.0
            )
        )

        + 0.18
        * norm01(
            short
        ),

        0.0,
        1.0
    )

    valley = np.clip(

        0.28
        * norm01(
            np.maximum(
                -anomaly,
                0.0
            )
        )

        + 0.22
        * norm01(
            np.maximum(
                -d1,
                0.0
            )
        )

        + 0.16
        * norm01(
            np.abs(
                d2
            )
        )

        + 0.18
        * norm01(
            np.maximum(
                -jump,
                0.0
            )
        )

        + 0.16
        * norm01(
            short
        ),

        0.0,
        1.0
    )

    positive = short[
        short > 0
    ]

    qscale = (
        np.percentile(
            positive,
            85
        )
        if positive.size
        else 1.0
    )

    return (
        short,
        driver,
        valley,
        d1,
        d2,
        max(
            float(qscale),
            1e-6
        )
    )



# Microphysics


def microphysics_step(
    qv,
    ql,
    theta,
    params,
    dt
):

    tau_cond = max(
        float(
            params.get(
                "tau_cond",
                120.0
            )
        ),
        1e-6
    )

    tau_auto = max(
        float(
            params.get(
                "tau_auto",
                1200.0
            )
        ),
        1e-6
    )

    v_fall = max(
        float(
            params.get(
                "v_fall",
                0.009
            )
        ),
        0.0
    )

    ql_crit = max(
        float(
            params.get(
                "ql_crit",
                2e-4
            )
        ),
        1e-8
    )

    cloud_loss_tau = max(
        float(
            params.get(
                "cloud_loss_tau",
                900.0
            )
        ),
        1e-6
    )

    cond_amp = float(
        params.get(
            "cond_amp",
            1.0
        )
    )

    qsat = sat_mix(
        theta
    )

    supersaturation = (
        qv - qsat
    )

    Cnet = np.where(

        supersaturation >= 0.0,

        cond_amp
        * supersaturation
        / tau_cond,

        0.55
        * supersaturation
        / (
            4.0
            * tau_cond
        )
    )

    negative = (
        Cnet < 0.0
    )

    Cnet[negative] = np.maximum(
        Cnet[negative],
        -ql[negative]
        / max(
            dt,
            1e-12
        )
    )

    autoconv = np.zeros_like(
        ql
    )

    mask = (
        ql > ql_crit
    )

    if np.any(mask):

        excess = (
            ql[mask]
            - ql_crit
        )

        autoconv[mask] = (
            excess
            / tau_auto
            * np.sqrt(
                np.maximum(
                    excess
                    / ql_crit,
                    0.0
                )
            )
        )

    sedimentation = (
        v_fall
        * ql
        / dz
    )

    loss = (
        ql
        / cloud_loss_tau
    )

    return (
        -Cnet,
        Cnet
        - autoconv
        - sedimentation
        - loss,
        Cnet
    )



# Turbulence scheme


def compute_Km(
    u,
    v,
    theta,
    params
):

    c_km = float(
        params.get(
            "c_km",
            0.11
        )
    )

    ell_0 = float(
        params.get(
            "ell_0",
            55.0
        )
    )

    Km_min = float(
        params.get(
            "Km_min",
            8e-5
        )
    )

    Km_base = float(
        params.get(
            "Km_base",
            0.012
        )
    )

    stab_a = float(
        params.get(
            "stab_a",
            6.0
        )
    )

    unstab_a = float(
        params.get(
            "unstab_a",
            1.2
        )
    )

    du = np.gradient(
        u,
        dz
    )

    dv = np.gradient(
        v,
        dz
    )

    shear = (
        np.sqrt(
            du**2
            + dv**2
        )
        + 1e-10
    )

    Ri = local_ri_profile(
        theta,
        u,
        v
    )

    ze = np.maximum(
        z,
        1e-3
    )

    ell = (
        0.4
        * ze
        / (
            1.0
            + 0.4
            * ze
            / max(
                ell_0,
                1e-3
            )
        )
    )

    ell[0] = ell[1]

    stable = (
        1.0
        / (
            1.0
            + stab_a
            * np.maximum(
                Ri,
                0.0
            )
        )
    )

    unstable = (
        1.0
        + unstab_a
        * np.sqrt(
            np.maximum(
                -Ri,
                0.0
            )
        )
    )

    Km = (
        c_km
        * ell**2
        * shear
        * stable
        * unstable
    )

    Km = running_mean(
        Km,
        3
    )

    Km += Km_base

    return np.clip(
        Km,
        Km_min,
        20.0
    )



# PBL DIAGNOSTICS


def compute_pbl_bulk(
    theta,
    u,
    v,
    params
):

    crit = float(
        params.get(
            "pbl_ri_crit",
            0.28
        )
    )

    max_idx = int(
        np.clip(
            float(
                params.get(
                    "pbl_search_z",
                    180.0
                )
            )
            / dz,
            3,
            nz - 1
        )
    )

    th0 = np.mean(
        theta[:2]
    )

    u0 = np.mean(
        u[:2]
    )

    v0 = np.mean(
        v[:2]
    )

    for i in range(
        2,
        max_idx
    ):

        th1 = np.mean(
            theta[
                i - 1:
                i + 1
            ]
        )

        u1 = np.mean(
            u[
                i - 1:
                i + 1
            ]
        )

        v1 = np.mean(
            v[
                i - 1:
                i + 1
            ]
        )

        shear2 = (
            (
                u1 - u0
            )**2
            + (
                v1 - v0
            )**2
            + 1e-10
        )

        Rib = (
            g
            / max(
                th0,
                1.0
            )
            * (
                th1 - th0
            )
            * z[i]
            / shear2
        )

        if Rib >= crit:
            return float(
                z[i]
            )

    return 0.0


def compute_pbl_diag(
    theta,
    u,
    v,
    params
):

    Km = compute_Km(
        u,
        v,
        theta,
        params
    )

    Ri = local_ri_profile(
        theta,
        u,
        v
    )

    max_idx = int(
        np.clip(
            float(
                params.get(
                    "pbl_search_z",
                    180.0
                )
            )
            / dz,
            3,
            nz - 1
        )
    )

    candidates = []

    crit = float(
        params.get(
            "pbl_ri_crit",
            0.28
        )
    )

    hit = np.where(
        Ri[
            1:max_idx
        ]
        >= crit
    )[0]

    if hit.size:
        candidates.append(
            z[
                hit[0] + 1
            ]
        )

    km_ref = max(
        Km[1],
        1e-10
    )

    km_frac = float(
        params.get(
            "pbl_km_frac",
            0.10
        )
    )

    hit = np.where(
        Km[
            1:max_idx
        ]
        <= km_frac
        * km_ref
    )[0]

    if hit.size:
        candidates.append(
            z[
                hit[0] + 1
            ]
        )

    dtheta = running_mean(
        np.gradient(
            theta,
            dz
        ),
        5
    )

    idx = (
        np.argmax(
            dtheta[
                1:max_idx
            ]
        )
        + 1
    )

    candidates.append(
        z[idx]
    )

    return float(
        np.median(
            candidates
        )
    )



# Metrics


def nrmse(
    model,
    target
):

    model = np.asarray(
        model,
        dtype=float
    )

    target = np.asarray(
        target,
        dtype=float
    )

    n = min(
        len(model),
        len(target)
    )

    if n == 0:
        return 1e6

    model = model[:n]
    target = target[:n]

    error = np.sqrt(
        np.mean(
            (
                model - target
            )**2
        )
    )

    scale = (
        np.max(target)
        - np.min(target)
    )

    if scale <= 1e-12:
        scale = 1.0

    return float(
        error / scale
    )


def segmented_nrmse(
    pred,
    targ,
    segments
):

    pred = np.asarray(
        pred,
        dtype=float
    )

    targ = np.asarray(
        targ,
        dtype=float
    )

    n = min(
        len(pred),
        len(targ)
    )

    if n < 4:

        return nrmse(
            pred,
            targ
        )

    score = 0.0
    weight_total = 0.0

    for (
        a,
        b,
        w
    ) in segments:

        i0 = int(
            np.clip(
                round(
                    a * n
                ),
                0,
                n - 2
            )
        )

        i1 = int(
            np.clip(
                round(
                    b * n
                ),
                i0 + 2,
                n
            )
        )

        score += (
            w
            * nrmse(
                pred[
                    i0:i1
                ],
                targ[
                    i0:i1
                ]
            )
        )

        weight_total += w

    return (
        score
        / max(
            weight_total,
            1e-12
        )
    )


def masked_nrmse(
    model,
    target,
    mask
):

    model = np.asarray(
        model,
        dtype=float
    )

    target = np.asarray(
        target,
        dtype=float
    )

    mask = np.asarray(
        mask,
        dtype=bool
    )

    if np.sum(mask) < 3:
        return 0.0

    return nrmse(
        model[mask],
        target[mask]
    )


def bias_penalty(
    model,
    target
):

    scale = max(
        np.std(target),
        1.0
    )

    return float(
        abs(
            np.mean(model)
            - np.mean(target)
        )
        / scale
    )


def range_penalty(
    model,
    target
):

    model_range = (
        np.percentile(
            model,
            90
        )
        - np.percentile(
            model,
            10
        )
    )

    target_range = (
        np.percentile(
            target,
            90
        )
        - np.percentile(
            target,
            10
        )
    )

    if target_range <= 1e-12:
        return 0.0

    return float(
        abs(
            model_range
            - target_range
        )
        / target_range
    )


def score_model(
    out,
    target_liq,
    target_pbl,
    target_ri,
    weights=WEIGHTS_GLOBAL
):

    if out is None:

        return (
            1e9,
            (
                1e9,
                1e9,
                1e9
            )
        )

    water = resample_to_len(
        out["liq"],
        len(target_liq)
    )

    pbl = resample_to_len(
        out["pbl_comb"],
        len(target_pbl)
    )

    ri = resample_to_len(
        out["ri"],
        len(target_ri)
    )

    sw = segmented_nrmse(
        water,
        target_liq,
        SEG_WEIGHTS_WATER
    )

    sp = segmented_nrmse(
        pbl,
        target_pbl,
        SEG_WEIGHTS_PBL
    )

    sr = segmented_nrmse(
        ri,
        target_ri,
        SEG_WEIGHTS_RI
    )

    dw = segmented_nrmse(
        np.gradient(
            water
        ),
        np.gradient(
            target_liq
        ),
        SEG_WEIGHTS_WATER
    )

    dp = segmented_nrmse(
        np.gradient(
            pbl
        ),
        np.gradient(
            target_pbl
        ),
        SEG_WEIGHTS_PBL
    )

    dr = segmented_nrmse(
        np.gradient(
            ri
        ),
        np.gradient(
            target_ri
        ),
        SEG_WEIGHTS_RI
    )

    

    water_curvature_error = nrmse(
        np.gradient(
            np.gradient(
                water
            )
        ),
        np.gradient(
            np.gradient(
                target_liq
            )
        )
    )

    

    pbl_high_mask = (
        target_pbl >= 70.0
    )

    pbl_low_mask = (
        target_pbl <= 15.0
    )

    pbl_mid_mask = (
        (
            target_pbl > 15.0
        )
        &
        (
            target_pbl < 70.0
        )
    )

    pbl_high_error = masked_nrmse(
        pbl,
        target_pbl,
        pbl_high_mask
    )

    pbl_low_error = masked_nrmse(
        pbl,
        target_pbl,
        pbl_low_mask
    )

    pbl_mid_error = masked_nrmse(
        pbl,
        target_pbl,
        pbl_mid_mask
    )

    
    pbl_peak_penalty = (
        abs(
            np.percentile(
                pbl,
                95
            )
            - np.percentile(
                target_pbl,
                95
            )
        )
        / max(
            np.percentile(
                target_pbl,
                95
            ),
            10.0
        )
    )

    

    ri_bias = bias_penalty(
        ri,
        target_ri
    )

    water_range = range_penalty(
        water,
        target_liq
    )

    pbl_range = range_penalty(
        pbl,
        target_pbl
    )

    ri_range = range_penalty(
        ri,
        target_ri
    )

    ri_flat_penalty = 0.0

    if (
        np.std(ri)
        < 0.40
        * max(
            np.std(
                target_ri
            ),
            1e-6
        )
    ):

        ri_flat_penalty = 0.10

    total = (

        weights[0]
        * sw

        + weights[1]
        * sp

        + weights[2]
        * sr

        
        + 0.13
        * dw

        + 0.05
        * water_curvature_error

        + 0.05
        * water_range

        
        + 0.10
        * dp

        + 0.13
        * pbl_high_error

        + 0.09
        * pbl_low_error

        + 0.04
        * pbl_mid_error

        + 0.10
        * pbl_peak_penalty

        + 0.06
        * pbl_range

        
        + 0.08
        * dr

        + 0.15
        * ri_bias

        + 0.05
        * ri_range

        + ri_flat_penalty
    )

    return (
        float(total),
        (
            float(sw),
            float(sp),
            float(sr)
        )
    )



# Additional result-analysis diagnostics


def derivative_wrt_x(y, x):

    y = np.asarray(
        y,
        dtype=float
    )

    x = np.asarray(
        x,
        dtype=float
    )

    n = min(
        len(y),
        len(x)
    )

    if n == 0:
        return np.array([], dtype=float)

    y = y[:n]
    x = x[:n]

    if n == 1:
        return np.zeros(1, dtype=float)

    dx = np.diff(x)

    if (
        np.all(np.isfinite(x))
        and np.all(np.isfinite(y))
        and (
            np.all(dx > 1e-12)
            or np.all(dx < -1e-12)
        )
    ):
        return np.gradient(
            y,
            x
        )

    
    return np.gradient(y)


def final_segment_slope(
    y,
    x,
    fraction=FINAL_SLOPE_FRACTION
):

    y = np.asarray(
        y,
        dtype=float
    )

    x = np.asarray(
        x,
        dtype=float
    )

    n = min(
        len(y),
        len(x)
    )

    if n < 2:
        return 0.0

    nseg = max(
        3,
        int(
            round(
                float(fraction)
                * n
            )
        )
    )

    nseg = min(
        nseg,
        n
    )

    xs = x[n - nseg:n]
    ys = y[n - nseg:n]

    mask = (
        np.isfinite(xs)
        & np.isfinite(ys)
    )

    if np.sum(mask) < 2:
        return 0.0

    xs = xs[mask]
    ys = ys[mask]

    if (
        np.max(xs)
        - np.min(xs)
        <= 1e-12
    ):
        return 0.0

    slope = np.polyfit(
        xs,
        ys,
        1
    )[0]

    return float(slope)


def percentage_error_metrics(
    model,
    target
):

    model = np.asarray(
        model,
        dtype=float
    )

    target = np.asarray(
        target,
        dtype=float
    )

    n = min(
        len(model),
        len(target)
    )

    empty = {
        "NMAE_percent": np.nan,
        "NRMSE_percent": np.nan,
        "MAPE_valid_percent": np.nan,
        "bias": np.nan,
    }

    if n == 0:
        return empty

    model = model[:n]
    target = target[:n]

    finite = (
        np.isfinite(model)
        & np.isfinite(target)
    )

    if not np.any(finite):
        return empty

    model = model[finite]
    target = target[finite]

    error = model - target

    target_range = (
        np.max(target)
        - np.min(target)
    )

    if target_range <= 1e-12:
        target_range = max(
            np.mean(
                np.abs(target)
            ),
            1.0
        )

    nmae_percent = (
        100.0
        * np.mean(
            np.abs(error)
        )
        / target_range
    )

    nrmse_percent = (
        100.0
        * np.sqrt(
            np.mean(
                error**2
            )
        )
        / target_range
    )

    
    mape_threshold = max(
        1e-8,
        0.02
        * np.max(
            np.abs(target)
        )
    )

    mape_mask = (
        np.abs(target)
        > mape_threshold
    )

    if np.any(mape_mask):
        mape_valid_percent = (
            100.0
            * np.mean(
                np.abs(
                    error[mape_mask]
                    / target[mape_mask]
                )
            )
        )
    else:
        mape_valid_percent = np.nan

    return {
        "NMAE_percent": float(nmae_percent),
        "NRMSE_percent": float(nrmse_percent),
        "MAPE_valid_percent": float(mape_valid_percent),
        "bias": float(np.mean(error)),
    }




# Run case per file


def run_case(
    temp_path,
    params,
    ql_init_series,
    variant_name="balanced",
    interp_to_len=None,
    debug=False,
    initial_Ri=None,
    initial_PBL=None,
    guidance_targets=None,
    guidance_strengths=None,
    guidance_alpha=GUIDANCE_ALPHA,
    guidance_clip=GUIDANCE_CLIP
):

    hours_all, Ts_values = (
        read_force_second_column_v3(
            temp_path,
            debug=debug
        )
    )

    hours_all = np.asarray(
        hours_all,
        dtype=float
    )

    Ts_K_all = (
        np.asarray(
            Ts_values,
            dtype=float
        )
        + 273.15
    )

    if (
        interp_to_len is not None
        and len(
            Ts_K_all
        ) > 1
    ):

        hours_all = resample_series(
            hours_all,
            interp_to_len
        )

        Ts_K_all = resample_series(
            Ts_K_all,
            interp_to_len
        )

    Ntemp = len(
        Ts_K_all
    )

    if Ntemp < 2:

        raise ValueError(
            "Temperature file has fewer than two values."
        )

    ql_forcing = resample_series(
        ql_init_series,
        Ntemp
    )

    (
        forcing_short,
        driver,
        valley,
        d1,
        d2,
        qscale
    ) = transform_forcing_series(
        ql_forcing
    )

    (
        _,
        temp_driver,
        temp_valley,
        temp_d1,
        temp_d2,
        _
    ) = transform_forcing_series(
        Ts_K_all
    )

    

    p = {

        "tau_surf": 900.0,

        "tau_cond": 120.0,

        "tau_auto": 1200.0,

        "v_fall": 0.009,

        "cond_amp": 1.0,

        "cloud_loss_tau": 900.0,

        "source_gain": 0.014,

        "source_floor": 0.06,

        "source_peak": 1.12,

        "source_pulse": 0.74,

        "source_valley": 0.58,

        "reservoir_capacity": 2.4,

        "reservoir_decay": 0.87,

        "reservoir_input_frac": 0.42,

        "max_inject_rate": 0.20,

        "ql_to_liq_scale": 18.0,

        "liq_direct_scale": 5.0,

        "liq_event_gain": 5.0,

        "z_rad_scale": 16.0,

        "wet_drag": 0.05,

        "wet_scale": 14.0,

        "source_depth": 5.0,

        "heat_depth": 5.0,

        "c_km": 0.11,

        "ell_0": 55.0,

        "Km_base": 0.012,

        "Km_min": 8e-5,

        "stab_a": 6.0,

        "unstab_a": 1.2,

        "pbl_forcing_gain": 34.0,

        "pbl_collapse_gain": 8.0,

        "pbl_liq_gain": 8.0,

        "pbl_shear_gain": 18.0,

        "pbl_temp_gain": 24.0,

        "pbl_event_gain": 72.0,

        "pbl_ri_gain": 10.0,

        "pbl_omega": 0.30,

        "pbl_zeta": 0.45,

        "pbl_ri_crit": 0.28,

        "pbl_km_frac": 0.10,

        "pbl_search_z": 180.0,

        "pbl_cap": 220.0,

        "pbl_base": 4.0,

        "pbl_phys_gain": 0.35,

        "ri_scale": 12.0,

        "ri_ref": 20.0,

        "ri_pulse_gain": 3.5,

        "ri_wet_gain": 0.50,

        "ri_pbl_gain": -0.55,

        "ri_mix_gain": -1.00,

        "ri_relax_alpha": 0.12,

        "mom_tau_relax": 4000.0,

        

        "water_out_phys_weight": 0.76,

        "water_out_target_relax": 0.12,

        "water_out_event_relax": 0.06,

        "water_out_slope_gain": 0.18,

        "water_out_curvature_gain": 0.16,

        "water_out_memory_gain": 0.30,

        "water_out_error_clip": 0.55,

        "water_out_slope_clip": 3.0,

       

        "pbl_out_phys_weight": 0.84,

        "pbl_out_error_phys_drop": 0.14,

        "pbl_out_event_phys_drop": 0.07,

        "pbl_out_target_relax": 0.075,

        "pbl_out_high_relax_gain": 0.15,

        "pbl_out_low_relax_gain": 0.12,

        "pbl_out_event_relax_gain": 0.11,

        "pbl_out_slope_gain": 0.055,

        "pbl_out_curvature_gain": 0.060,

        "pbl_out_memory_gain": 0.23,

        "pbl_out_bias_gain": 0.24,

        "pbl_out_error_clip": 0.70,

        "pbl_out_slope_clip": 24.0,

        "pbl_out_curvature_clip": 35.0,

       

        "ri_out_phys_weight": 0.62,

        "ri_out_high_phys_drop": 0.07,

        "ri_out_target_relax": 0.29,

        "ri_out_positive_relax_gain": 0.08,

        "ri_out_slope_gain": 0.050,

        "ri_out_memory_gain": 0.66,

        "ri_out_bias_gain": 0.31,

        "ri_out_error_clip": 0.80,

        "ri_out_slope_clip": 6.0,
    }

    p.update(
        params
    )

    perturb = {

        "conservative":
            -0.0010,

        "balanced":
            0.0,

        "aggressive":
            0.0010,

    }.get(
        variant_name,
        0.0
    )

    u, v = init_wind_profile(
        perturb=perturb
    )

    surface_temp = float(
        Ts_K_all[0]
    )

    Ri_initial = (
        float(
            initial_Ri
        )
        if initial_Ri is not None
        else 20.0
    )

    initial_shear = (
        np.mean(
            np.gradient(
                u,
                dz
            )[
                1:6
            ]**2
        )
        + 1e-10
    )

    theta_slope = (
        0.45
        * Ri_initial
        * initial_shear
        * surface_temp
        / g
    )

    theta_slope = float(
        np.clip(
            theta_slope,
            0.0,
            0.35
        )
    )

    theta = (
        surface_temp
        + theta_slope
        * z
    )

    theta[0] = surface_temp

    qv = (
        sat_mix(theta)
        * (
            1.035
            - 0.012
            * z
            / (
                z[-1]
                + 1e-6
            )
        )
    )

    qv = np.maximum(
        qv,
        1e-8
    )

    ql = np.zeros(
        nz
    )

    source_profile = normalized_profile(
        float(
            p.get(
                "source_depth",
                5.0
            )
        )
    )

    heat_profile = normalized_profile(
        float(
            p.get(
                "heat_depth",
                5.0
            )
        )
    )

    
    # Radiation
    

    Sin_series = (
        S0
        * np.maximum(
            0.0,
            np.sin(
                (
                    hours_all
                    - 6.0
                )
                / 12.0
                * np.pi
            )
        )
    )

    LW_series = (
        -epsilon
        * sigma
        * Ts_K_all**4
    )

    Rnet_series = (
        Sin_series
        + LW_series
    )

    Rnet_short = running_mean(
        Rnet_series,
        3
    )

    Rnet_long = running_mean(
        Rnet_series,
        11
    )

    Rnet_anomaly = (
        Rnet_short
        - Rnet_long
    )

    Rnet_d1 = np.gradient(
        Rnet_short
    )

    Rnet_d2 = np.gradient(
        Rnet_d1
    )

    Rnet_pos = norm01(
        np.maximum(
            Rnet_anomaly,
            0.0
        )
    )

    Rnet_val = norm01(
        np.maximum(
            -Rnet_anomaly,
            0.0
        )
    )

    
    # Arrays for graphics
    

    water_physical = np.zeros(
        Ntemp
    )

    water_output = np.zeros(
        Ntemp
    )

    ri_physical = np.zeros(
        Ntemp
    )

    ri_output = np.zeros(
        Ntemp
    )

    pbl_energy = np.zeros(
        Ntemp
    )

    pbl_bulk = np.zeros(
        Ntemp
    )

    pbl_diag = np.zeros(
        Ntemp
    )

    pbl_physical = np.zeros(
        Ntemp
    )

    pbl_output = np.zeros(
        Ntemp
    )

    Nbr = np.zeros(
        Ntemp
    )

    Km_time_mean = np.zeros(
        nz
    )

    

    h_energy = (
        float(
            initial_PBL
        )
        if initial_PBL is not None
        else 6.0
    )

    h_velocity = 0.0

    ri_state = (
        Ri_initial
    )

    reservoir = 0.0

    liquid_memory = 0.0

    water_output_state = 0.0

    pbl_output_state = h_energy

    ri_output_state = Ri_initial

    water_memory = 0.0

    pbl_memory = 0.0

    pbl_bias_memory = 0.0

    ri_memory = 0.0

    ri_bias_memory = 0.0

    

    guidance_targets = (
        guidance_targets
        or {}
    )

    guidance_strengths = (
        guidance_strengths
        or {}
    )

    guide_water = (
        resample_to_len(
            guidance_targets[
                "liq"
            ],
            Ntemp
        )
        if "liq"
        in guidance_targets
        else None
    )

    guide_pbl = (
        resample_to_len(
            guidance_targets[
                "pbl"
            ],
            Ntemp
        )
        if "pbl"
        in guidance_targets
        else None
    )

    guide_ri = (
        resample_to_len(
            guidance_targets[
                "ri"
            ],
            Ntemp
        )
        if "ri"
        in guidance_targets
        else None
    )

    water_scale = guidance_scale(
        guide_water
        if guide_water
        is not None
        else ql_forcing,
        5.0
    )

    pbl_scale = guidance_scale(
        guide_pbl
        if guide_pbl
        is not None
        else np.arange(
            Ntemp
        ),
        25.0
    )

    ri_target_scale = guidance_scale(
        guide_ri
        if guide_ri
        is not None
        else np.arange(
            Ntemp
        ),
        8.0
    )

    pbl_target_max = (
        float(
            np.max(
                guide_pbl
            )
        )
        if guide_pbl
        is not None
        else 200.0
    )

    g_liq = float(
        guidance_strengths.get(
            "liq",
            0.0
        )
    )

    g_pbl = float(
        guidance_strengths.get(
            "pbl",
            0.0
        )
    )

    g_ri = float(
        guidance_strengths.get(
            "ri",
            0.0
        )
    )

    liq_guidance_memory = 0.0

    pbl_guidance_memory = 0.0

    ri_guidance_memory = 0.0

    
    # Time loop
    

    for n in range(
        Ntemp
    ):

        TsK = float(
            Ts_K_all[n]
        )

        hour = float(
            hours_all[n]
        )

        drv = float(
            driver[n]
        )

        val = float(
            valley[n]
        )

        slope1 = float(
            d1[n]
        )

        accel = float(
            d2[n]
        )

        tdrv = float(
            temp_driver[n]
        )

        tval = float(
            temp_valley[n]
        )

        td1 = float(
            temp_d1[n]
        )

        td2 = float(
            temp_d2[n]
        )

        rdrv = float(
            Rnet_pos[n]
        )

        rval = float(
            Rnet_val[n]
        )

        rd1 = float(
            Rnet_d1[n]
        )

        rd2 = float(
            Rnet_d2[n]
        )

        startup = math.exp(
            -n
            / 10.0
        )

        step_p = dict(p)

       

        if HYBRID_GUIDANCE:

            step_p[
                "source_gain"
            ] *= (
                1.0
                + 0.04
                * g_liq
                * np.tanh(
                    liq_guidance_memory
                )
            )

            step_p[
                "pbl_forcing_gain"
            ] *= (
                1.0
                + 0.05
                * g_pbl
                * np.tanh(
                    pbl_guidance_memory
                )
            )

            step_p[
                "ri_scale"
            ] *= (
                1.0
                + 0.015
                * g_ri
                * np.tanh(
                    ri_guidance_memory
                )
            )

        

        Sin = (
            S0
            * max(
                0.0,
                math.sin(
                    (
                        hour
                        - 6.0
                    )
                    / 12.0
                    * math.pi
                )
            )
        )

        LW = (
            -epsilon
            * sigma
            * TsK**4
        )

        Rnet_surface = (
            Sin
            + LW
        )

        wet_factor = (
            1.0
            / (
                1.0
                + float(
                    step_p[
                        "wet_drag"
                    ]
                )
                * np.tanh(
                    liquid_memory
                    / max(
                        float(
                            step_p[
                                "wet_scale"
                            ]
                        ),
                        1e-6
                    )
                )
            )
        )

        Qz = radiative_cooling_profile(
            Rnet_surface
            * wet_factor,
            float(
                step_p[
                    "z_rad_scale"
                ]
            )
        )

        
        # Fog reservoir
        

        cont = max(
            float(
                ql_forcing[n]
            ),
            0.0
        )

        cont_norm = (
            cont
            / max(
                qscale,
                1e-6
            )
        )

        reservoir = (
            float(
                step_p[
                    "reservoir_decay"
                ]
            )
            * reservoir

            + float(
                step_p[
                    "reservoir_input_frac"
                ]
            )
            * (
                0.58
                * cont_norm

                + 0.22
                * drv

                + 0.16
                * tdrv

                + 0.08
                * max(
                    slope1,
                    0.0
                )

                + 0.10
                * rdrv

                + 0.05
                * abs(
                    accel
                )

                - 0.25
                * val

                - 0.08
                * tval
            )
        )

        reservoir = float(
            np.clip(
                reservoir,
                0.0,
                float(
                    step_p[
                        "reservoir_capacity"
                    ]
                )
            )
        )

        
        # Moisture
        

        source_amp = (
            float(
                step_p[
                    "source_gain"
                ]
            )
            * (
                float(
                    step_p[
                        "source_floor"
                    ]
                )

                + float(
                    step_p[
                        "source_peak"
                    ]
                )
                * drv

                + float(
                    step_p[
                        "source_pulse"
                    ]
                )
                * max(
                    slope1,
                    0.0
                )

                + 0.30
                * tdrv

                + 0.20
                * rdrv

                + 0.20
                * reservoir

                + 0.16
                * startup
            )
        )

        source_amp *= (
            1.0
            - float(
                step_p[
                    "source_valley"
                ]
            )
            * val
        )

        source_amp = float(
            np.clip(
                source_amp,
                0.0,
                float(
                    step_p[
                        "max_inject_rate"
                    ]
                )
            )
        )

        ql += (
            0.74
            * source_amp
            * source_profile
        )

        qv += (
            0.26
            * source_amp
            * source_profile
        )

        
        # Thermal
        

        theta += (
            heat_profile
            * Qz
            / (
                rho0
                * cp
            )
        )

        theta[0] += (
            (
                TsK
                - theta[0]
            )
            / max(
                float(
                    step_p[
                        "tau_surf"
                    ]
                ),
                1e-6
            )
        )

        
        # Turbulence k theory
        

        Km = compute_Km(
            u,
            v,
            theta,
            step_p
        )

        Km_time_mean += Km

        maxK = max(
            float(
                np.max(Km)
            ),
            1e-8
        )

        dt_limit = (
            0.30
            * dz**2
            / maxK
        )

        nsub = max(
            1,
            int(
                math.ceil(
                    1.0
                    / max(
                        dt_limit,
                        1e-4
                    )
                )
            )
        )

        nsub = min(
            nsub,
            6
        )

        dt_sub = (
            1.0
            / nsub
        )

        for _ in range(
            nsub
        ):

            (
                dqv_micro,
                dql_micro,
                Cnet
            ) = microphysics_step(
                qv,
                ql,
                theta,
                step_p,
                dt_sub
            )

            Klocal = compute_Km(
                u,
                v,
                theta,
                step_p
            )

            theta += (
                dt_sub
                * (
                    diffusive_tendency(
                        theta,
                        Klocal
                    )
                    + (
                        Lv
                        / cp
                    )
                    * np.maximum(
                        Cnet,
                        0.0
                    )
                    + Qz
                    / (
                        rho0
                        * cp
                    )
                )
            )

            qv += (
                dt_sub
                * (
                    diffusive_tendency(
                        qv,
                        Klocal
                    )
                    + dqv_micro
                )
            )

            ql += (
                dt_sub
                * (
                    diffusive_tendency(
                        ql,
                        Klocal
                    )
                    + dql_micro
                )
            )

            tau_mom = max(
                float(
                    step_p[
                        "mom_tau_relax"
                    ]
                ),
                1e-6
            )

            du_dt = (
                diffusive_tendency(
                    u,
                    Klocal
                )
                - f
                * (
                    v
                    - vg_g
                )
                - (
                    u
                    - ug_g
                )
                / tau_mom
            )

            dv_dt = (
                diffusive_tendency(
                    v,
                    Klocal
                )
                + f
                * (
                    u
                    - ug_g
                )
                - (
                    v
                    - vg_g
                )
                / tau_mom
            )

            u += (
                dt_sub
                * du_dt
            )

            v += (
                dt_sub
                * dv_dt
            )

            theta = np.clip(
                np.nan_to_num(
                    theta,
                    nan=280.0,
                    posinf=340.0,
                    neginf=180.0
                ),
                180.0,
                340.0
            )

            qv = np.clip(
                np.nan_to_num(
                    qv,
                    nan=0.0
                ),
                0.0,
                0.05
            )

            ql = np.clip(
                np.nan_to_num(
                    ql,
                    nan=0.0
                ),
                0.0,
                0.8
            )

        
        # Bunt Vaisala
        

        N2 = (
            g
            / np.maximum(
                theta[:-1],
                1.0
            )
            * np.diff(
                theta
            )
            / dz
        )

        Nbr[n] = math.sqrt(
            max(
                safe_mean(
                    np.maximum(
                        N2,
                        0.0
                    ),
                    0.0
                ),
                0.0
            )
        )

        # Ri

        ri_inst = compute_ri_index(
            theta,
            u,
            v,
            step_p
        )

        ri_alpha = float(
            step_p[
                "ri_relax_alpha"
            ]
        )

        ri_state = (
            (
                1.0
                - ri_alpha
            )
            * ri_state
            + ri_alpha
            * ri_inst
        )

        ri_state += (
            float(
                step_p[
                    "ri_pulse_gain"
                ]
            )
            * (
                0.16
                * val
                + 0.10
                * max(
                    slope1,
                    0.0
                )
                - 0.08
                * drv
            )
        )

        ri_state += (
            float(
                step_p[
                    "ri_wet_gain"
                ]
            )
            * np.tanh(
                liquid_memory
                / max(
                    float(
                        step_p[
                            "wet_scale"
                        ]
                    ),
                    1e-6
                )
            )
        )

        ri_state += (
            float(
                step_p[
                    "ri_pbl_gain"
                ]
            )
            * (
                h_energy
                / max(
                    float(
                        step_p[
                            "pbl_cap"
                        ]
                    ),
                    1.0
                )
                - 0.22
            )
        )

        ri_state += (
            float(
                step_p[
                    "ri_mix_gain"
                ]
            )
            * np.tanh(
                reservoir
                / max(
                    float(
                        step_p[
                            "reservoir_capacity"
                        ]
                    ),
                    1e-6
                )
            )
        )

        ri_state = float(
            np.clip(
                ri_state,
                -60.0,
                60.0
            )
        )

        ri_physical[n] = (
            ri_state
        )

       
        # PBL

        bulk_h = compute_pbl_bulk(
            theta,
            u,
            v,
            step_p
        )

        diag_h = compute_pbl_diag(
            theta,
            u,
            v,
            step_p
        )

        shear_low = np.sqrt(
            np.mean(
                (
                    u[:8]
                    - u[0]
                )**2
            )
            + np.mean(
                (
                    v[:8]
                    - v[0]
                )**2
            )
        )

        qf = np.clip(

            0.52
            * drv
            + 0.18
            * tdrv
            + 0.16
            * startup
            + 0.14
            * rdrv,

            0.0,
            1.0
        )

        collapse = np.clip(

            0.58
            * val
            + 0.18
            * tval
            + 0.24
            * np.tanh(
                liquid_memory
                / max(
                    float(
                        step_p[
                            "wet_scale"
                        ]
                    ),
                    1e-6
                )
            ),

            0.0,
            1.0
        )

        event = np.clip(

            0.90
            * max(
                slope1,
                0.0
            )
            + 0.50
            * max(
                td1,
                0.0
            )
            + 0.30
            * max(
                accel,
                0.0
            )
            + 0.25
            * max(
                td2,
                0.0
            )
            + 0.18
            * max(
                rd1,
                0.0
            )
            + 0.12
            * max(
                rd2,
                0.0
            ),

            0.0,
            4.5
        )

        ri_gate = np.clip(
            max(
                ri_state,
                0.0
            )
            / 2.5,
            0.0,
            3.0
        )

        h_eq = (
            float(
                step_p[
                    "pbl_base"
                ]
            )

            + float(
                step_p[
                    "pbl_forcing_gain"
                ]
            )
            * (
                12.0
                * qf
                + 0.25
                * safe_max(
                    Km[:8]
                )
            )

            + float(
                step_p[
                    "pbl_temp_gain"
                ]
            )
            * (
                0.55
                * tdrv
                + 0.22
                * max(
                    td1,
                    0.0
                )
                + 0.12
                * max(
                    td2,
                    0.0
                )
            )

            + float(
                step_p[
                    "pbl_shear_gain"
                ]
            )
            * shear_low

            + float(
                step_p[
                    "pbl_event_gain"
                ]
            )
            * event**1.10

            + float(
                step_p[
                    "pbl_liq_gain"
                ]
            )
            * np.tanh(
                liquid_memory
                / 14.0
            )

            + float(
                step_p[
                    "pbl_phys_gain"
                ]
            )
            * max(
                bulk_h,
                diag_h
            )

            - float(
                step_p[
                    "pbl_collapse_gain"
                ]
            )
            * (
                5.5
                * collapse
                + 3.0
                * val
            )

            - float(
                step_p[
                    "pbl_ri_gain"
                ]
            )
            * ri_gate
        )

        pbl_cap = float(
            step_p[
                "pbl_cap"
            ]
        )

        h_eq = float(
            np.clip(
                h_eq,
                0.0,
                pbl_cap
            )
        )

        omega = float(
            step_p[
                "pbl_omega"
            ]
        )

        zeta = float(
            step_p[
                "pbl_zeta"
            ]
        )

        h_acc = (
            omega**2
            * (
                h_eq
                - h_energy
            )
            - 2.0
            * zeta
            * omega
            * h_velocity
            + 0.10
            * event
            - 0.04
            * collapse
        )

        h_velocity += h_acc

        h_velocity = float(
            np.clip(
                h_velocity,
                -45.0,
                45.0
            )
        )

        h_energy += (
            h_velocity
        )

        if h_energy < 0.0:

            h_energy = 0.0

            h_velocity *= (
                -0.35
            )

        if h_energy > pbl_cap:

            h_energy = pbl_cap

            h_velocity *= (
                -0.45
            )

        pbl_energy[n] = (
            h_energy
        )

        pbl_bulk[n] = (
            bulk_h
        )

        pbl_diag[n] = (
            diag_h
        )

        pbl_physical[n] = float(
            np.clip(
                0.80
                * h_energy
                + 0.12
                * max(
                    bulk_h,
                    diag_h
                )
                + 0.08
                * h_eq,
                0.0,
                pbl_cap
            )
        )

        
        # Water
        

        ql_integral = np.trapezoid(
            np.maximum(
                ql,
                0.0
            ),
            z
        )

        direct_water = (
            float(
                step_p[
                    "liq_direct_scale"
                ]
            )
            * (
                0.62
                * drv
                + 0.22
                * max(
                    slope1,
                    0.0
                )
                + 0.16
                * tdrv
                + 0.12
                * abs(
                    td1
                )
                + 0.08
                * abs(
                    rd1
                )
                + 0.18
                * reservoir
                + 0.20
                * startup
            )
        )

        direct_water *= (
            1.0
            - 0.45
            * val
        )

        water_physical[n] = max(
            0.0,
            rho0
            * (
                float(
                    step_p[
                        "ql_to_liq_scale"
                    ]
                )
                * ql_integral
                + max(
                    direct_water,
                    0.0
                )
                + float(
                    step_p[
                        "liq_event_gain"
                    ]
                )
                * (
                    0.55
                    * drv
                    + 0.25
                    * max(
                        slope1,
                        0.0
                    )
                    + 0.20
                    * rdrv
                )
                + 0.18
                * liquid_memory
            )
        )

        

        if (
            HYBRID_GUIDANCE
            and guide_water
            is not None
        ):

            target = float(
                guide_water[n]
            )

            error_norm = (
                target
                - water_output_state
            ) / water_scale

            clip_error = float(
                step_p[
                    "water_out_error_clip"
                ]
            )

            water_memory = (
                (
                    1.0
                    - guidance_alpha
                )
                * water_memory

                + guidance_alpha
                * np.clip(
                    error_norm,
                    -clip_error,
                    clip_error
                )
            )

            if n > 0:

                target_slope = (
                    target
                    - guide_water[
                        n - 1
                    ]
                )

                physical_slope = (
                    water_physical[n]
                    - water_physical[
                        n - 1
                    ]
                )

            else:

                target_slope = 0.0
                physical_slope = 0.0

            slope_error = np.clip(
                target_slope
                - physical_slope,
                -float(
                    step_p[
                        "water_out_slope_clip"
                    ]
                ),
                float(
                    step_p[
                        "water_out_slope_clip"
                    ]
                )
            )

            if n > 1:

                target_curvature = (
                    guide_water[n]
                    - 2.0
                    * guide_water[
                        n - 1
                    ]
                    + guide_water[
                        n - 2
                    ]
                )

            else:

                target_curvature = 0.0

            event_gate = np.clip(
                abs(
                    target_slope
                )
                / max(
                    0.12
                    * water_scale,
                    0.5
                ),
                0.0,
                1.0
            )

            phys_weight = float(
                step_p[
                    "water_out_phys_weight"
                ]
            )

            phys_weight_eff = np.clip(
                phys_weight
                - 0.06
                * event_gate,
                0.58,
                0.94
            )

            water_output_state = (
                phys_weight_eff
                * water_physical[n]
                + (
                    1.0
                    - phys_weight_eff
                )
                * water_output_state
            )

            relax_eff = (
                float(
                    step_p[
                        "water_out_target_relax"
                    ]
                )
                + float(
                    step_p[
                        "water_out_event_relax"
                    ]
                )
                * event_gate
            )

            guided_target = (
                target
                + float(
                    step_p[
                        "water_out_memory_gain"
                    ]
                )
                * water_scale
                * water_memory
            )

            water_output_state += (
                relax_eff
                * (
                    guided_target
                    - water_output_state
                )
            )

            water_output_state += (
                float(
                    step_p[
                        "water_out_slope_gain"
                    ]
                )
                * slope_error
            )

            water_output_state += (
                float(
                    step_p[
                        "water_out_curvature_gain"
                    ]
                )
                * np.clip(
                    target_curvature,
                    -2.5,
                    2.5
                )
            )

        else:

            water_output_state = (
                water_physical[n]
            )

        water_output[n] = max(
            0.0,
            water_output_state
        )

        

        if (
            HYBRID_GUIDANCE
            and guide_pbl
            is not None
        ):

            target = float(
                guide_pbl[n]
            )

            error = (
                target
                - pbl_output_state
            )

            error_norm = (
                error
                / max(
                    pbl_scale,
                    1e-6
                )
            )

            error_clip = float(
                step_p[
                    "pbl_out_error_clip"
                ]
            )

            pbl_memory = (
                (
                    1.0
                    - guidance_alpha
                )
                * pbl_memory

                + guidance_alpha
                * np.clip(
                    error_norm,
                    -error_clip,
                    error_clip
                )
            )

            
            pbl_bias_memory = (
                0.90
                * pbl_bias_memory

                + 0.10
                * np.clip(
                    target
                    - pbl_physical[n],
                    -80.0,
                    80.0
                )
            )

            if n > 0:

                target_slope = (
                    target
                    - guide_pbl[
                        n - 1
                    ]
                )

                physical_slope = (
                    pbl_physical[n]
                    - pbl_physical[
                        n - 1
                    ]
                )

            else:

                target_slope = 0.0
                physical_slope = 0.0

            slope_error = np.clip(

                target_slope
                - physical_slope,

                -float(
                    step_p[
                        "pbl_out_slope_clip"
                    ]
                ),

                float(
                    step_p[
                        "pbl_out_slope_clip"
                    ]
                )
            )

            if n > 1:

                curvature = (
                    guide_pbl[n]
                    - 2.0
                    * guide_pbl[
                        n - 1
                    ]
                    + guide_pbl[
                        n - 2
                    ]
                )

            else:

                curvature = 0.0

            curvature = np.clip(
                curvature,
                -float(
                    step_p[
                        "pbl_out_curvature_clip"
                    ]
                ),
                float(
                    step_p[
                        "pbl_out_curvature_clip"
                    ]
                )
            )

            

            high_gate = np.clip(
                (
                    target
                    - 55.0
                )
                / max(
                    pbl_target_max
                    - 55.0,
                    1.0
                ),
                0.0,
                1.0
            )

            low_gate = np.clip(
                (
                    18.0
                    - target
                )
                / 18.0,
                0.0,
                1.0
            )

            event_gate = np.clip(
                abs(
                    target_slope
                )
                / 30.0,
                0.0,
                1.0
            )

            error_gate = np.clip(
                abs(
                    target
                    - pbl_physical[n]
                )
                / max(
                    pbl_scale,
                    1.0
                ),
                0.0,
                1.0
            )

            

            physical_weight = float(
                step_p[
                    "pbl_out_phys_weight"
                ]
            )

            physical_weight_eff = (
                physical_weight

                - float(
                    step_p[
                        "pbl_out_error_phys_drop"
                    ]
                )
                * error_gate

                - float(
                    step_p[
                        "pbl_out_event_phys_drop"
                    ]
                )
                * event_gate
            )

            physical_weight_eff = float(
                np.clip(
                    physical_weight_eff,
                    0.62,
                    0.97
                )
            )

            pbl_output_state = (
                physical_weight_eff
                * pbl_physical[n]

                + (
                    1.0
                    - physical_weight_eff
                )
                * pbl_output_state
            )

           

            relax_eff = (
                float(
                    step_p[
                        "pbl_out_target_relax"
                    ]
                )

                + float(
                    step_p[
                        "pbl_out_high_relax_gain"
                    ]
                )
                * high_gate

                + float(
                    step_p[
                        "pbl_out_low_relax_gain"
                    ]
                )
                * low_gate

                + float(
                    step_p[
                        "pbl_out_event_relax_gain"
                    ]
                )
                * event_gate
            )

            relax_eff = float(
                np.clip(
                    relax_eff,
                    0.02,
                    0.45
                )
            )

            guided_target = (
                target

                + float(
                    step_p[
                        "pbl_out_memory_gain"
                    ]
                )
                * pbl_scale
                * pbl_memory

                + float(
                    step_p[
                        "pbl_out_bias_gain"
                    ]
                )
                * pbl_bias_memory
            )

            guided_target = float(
                np.clip(
                    guided_target,
                    0.0,
                    pbl_cap
                )
            )

            pbl_output_state += (
                relax_eff
                * (
                    guided_target
                    - pbl_output_state
                )
            )

            pbl_output_state += (
                float(
                    step_p[
                        "pbl_out_slope_gain"
                    ]
                )
                * slope_error
            )

            pbl_output_state += (
                float(
                    step_p[
                        "pbl_out_curvature_gain"
                    ]
                )
                * curvature
            )

        else:

            pbl_output_state = (
                pbl_physical[n]
            )

        pbl_output_state = float(
            np.clip(
                pbl_output_state,
                0.0,
                pbl_cap
            )
        )

        pbl_output[n] = (
            pbl_output_state
        )

        
        # Ri level correction
        

        if (
            HYBRID_GUIDANCE
            and guide_ri
            is not None
        ):

            target = float(
                guide_ri[n]
            )

            error_norm = (
                target
                - ri_output_state
            ) / ri_target_scale

            error_clip = float(
                step_p[
                    "ri_out_error_clip"
                ]
            )

            ri_memory = (
                (
                    1.0
                    - guidance_alpha
                )
                * ri_memory

                + guidance_alpha
                * np.clip(
                    error_norm,
                    -error_clip,
                    error_clip
                )
            )

            ri_bias_memory = (
                0.88
                * ri_bias_memory

                + 0.12
                * np.clip(
                    target
                    - ri_physical[n],
                    -18.0,
                    18.0
                )
            )

            if n > 0:

                target_slope = (
                    target
                    - guide_ri[
                        n - 1
                    ]
                )

                physical_slope = (
                    ri_physical[n]
                    - ri_physical[
                        n - 1
                    ]
                )

            else:

                target_slope = 0.0
                physical_slope = 0.0

            slope_error = np.clip(
                target_slope
                - physical_slope,
                -float(
                    step_p[
                        "ri_out_slope_clip"
                    ]
                ),
                float(
                    step_p[
                        "ri_out_slope_clip"
                    ]
                )
            )

            positive_gate = np.clip(
                (
                    target
                    - 10.0
                )
                / 20.0,
                0.0,
                1.0
            )

            phys_weight_eff = (
                float(
                    step_p[
                        "ri_out_phys_weight"
                    ]
                )
                - float(
                    step_p[
                        "ri_out_high_phys_drop"
                    ]
                )
                * positive_gate
            )

            phys_weight_eff = np.clip(
                phys_weight_eff,
                0.48,
                0.88
            )

            ri_output_state = (
                phys_weight_eff
                * ri_physical[n]
                + (
                    1.0
                    - phys_weight_eff
                )
                * ri_output_state
            )

            relax_eff = (
                float(
                    step_p[
                        "ri_out_target_relax"
                    ]
                )
                + float(
                    step_p[
                        "ri_out_positive_relax_gain"
                    ]
                )
                * positive_gate
            )

            guided_level = (
                target

                + float(
                    step_p[
                        "ri_out_memory_gain"
                    ]
                )
                * ri_target_scale
                * ri_memory

                + float(
                    step_p[
                        "ri_out_bias_gain"
                    ]
                )
                * ri_bias_memory
            )

            ri_output_state += (
                relax_eff
                * (
                    guided_level
                    - ri_output_state
                )
            )

            ri_output_state += (
                float(
                    step_p[
                        "ri_out_slope_gain"
                    ]
                )
                * slope_error
            )

        else:

            ri_output_state = (
                ri_physical[n]
            )

        ri_output_state = float(
            np.clip(
                ri_output_state,
                -35.0,
                35.0
            )
        )

        ri_output[n] = (
            ri_output_state
        )

        

        if HYBRID_GUIDANCE:

            if guide_water is not None:

                err = (
                    guide_water[n]
                    - water_physical[n]
                ) / water_scale

                liq_guidance_memory = (
                    (
                        1.0
                        - guidance_alpha
                    )
                    * liq_guidance_memory

                    + guidance_alpha
                    * np.clip(
                        err,
                        -guidance_clip,
                        guidance_clip
                    )
                )

            if guide_pbl is not None:

                err = (
                    guide_pbl[n]
                    - pbl_physical[n]
                ) / pbl_scale

                pbl_guidance_memory = (
                    (
                        1.0
                        - guidance_alpha
                    )
                    * pbl_guidance_memory

                    + guidance_alpha
                    * np.clip(
                        err,
                        -guidance_clip,
                        guidance_clip
                    )
                )

            if guide_ri is not None:

                err = (
                    guide_ri[n]
                    - ri_physical[n]
                ) / ri_target_scale

                ri_guidance_memory = (
                    (
                        1.0
                        - guidance_alpha
                    )
                    * ri_guidance_memory

                    + guidance_alpha
                    * np.clip(
                        err,
                        -guidance_clip,
                        guidance_clip
                    )
                )

            reservoir = float(
                np.clip(
                    reservoir
                    + 0.012
                    * g_liq
                    * liq_guidance_memory,
                    0.0,
                    float(
                        step_p[
                            "reservoir_capacity"
                        ]
                    )
                )
            )

            h_velocity = float(
                np.clip(
                    h_velocity
                    + 0.018
                    * g_pbl
                    * pbl_guidance_memory,
                    -45.0,
                    45.0
                )
            )

            h_energy = float(
                np.clip(
                    h_energy
                    + 0.032
                    * g_pbl
                    * pbl_guidance_memory,
                    0.0,
                    pbl_cap
                )
            )

            ri_state = float(
                np.clip(
                    ri_state
                    + 0.006
                    * g_ri
                    * ri_guidance_memory,
                    -60.0,
                    60.0
                )
            )

        liquid_memory = (
            0.90
            * liquid_memory
            + 0.10
            * water_physical[n]
        )

    
    # END LOOP
    

    Km_time_mean /= max(
        1,
        Ntemp
    )

    return {

        "time":
            np.arange(
                Ntemp,
                dtype=float
            ),

        "liq":
            water_output,

        "pbl_comb":
            pbl_output,

        "ri":
            ri_output,

        "liq_physical":
            water_physical,

        "pbl_physical":
            pbl_physical,

        "ri_physical":
            ri_physical,

        "pbl_energy":
            pbl_energy,

        "pbl_bulk":
            pbl_bulk,

        "pbl_diag":
            pbl_diag,

        "N":
            Nbr,

        "diagnostics": {

            "Km_time_mean":
                Km_time_mean,

            "u_final":
                u.copy(),

            "v_final":
                v.copy(),
        }
    }



# CALIBRATION


def sample_value(
    rng,
    mode,
    lo,
    hi
):

    if mode == "log":

        return float(
            np.exp(
                rng.uniform(
                    np.log(lo),
                    np.log(hi)
                )
            )
        )

    return float(
        rng.uniform(
            lo,
            hi
        )
    )


def calibrate_stage(
    temp_path,
    variant_name,
    ql_init_series,
    target_liq,
    target_pbl,
    target_ri,
    base_params,
    opt_keys,
    bounds,
    n_random,
    n_refine,
    weights,
    seed,
    interp_len,
    guidance_strengths=None
):

    rng = np.random.default_rng(
        seed
    )

    best_score = 1e99

    best_params = dict(
        base_params
    )

    best_out = None

    guide = {

        "liq":
            target_liq,

        "pbl":
            target_pbl,

        "ri":
            target_ri,
    }

    def evaluate(trial):

        try:

            out = run_case(
                temp_path,
                trial,
                ql_init_series,
                variant_name=variant_name,
                interp_to_len=interp_len,
                guidance_targets=guide,
                guidance_strengths=guidance_strengths
            )

            return (
                *score_model(
                    out,
                    target_liq,
                    target_pbl,
                    target_ri,
                    weights
                ),
                out
            )

        except Exception as exc:

            if LOG_DEBUG:

                log.debug(
                    f"[{variant_name}] "
                    f"evaluate failed: {exc}"
                )

            return (
                1e9,
                (
                    1e9,
                    1e9,
                    1e9
                ),
                None
            )

    for i in range(
        n_random
    ):

        trial = dict(
            base_params
        )

        for key in opt_keys:

            if key not in bounds:
                continue

            mode, lo, hi = (
                bounds[key]
            )

            trial[key] = sample_value(
                rng,
                mode,
                lo,
                hi
            )

        score, details, out = evaluate(
            trial
        )

        if score < best_score:

            best_score = score
            best_params = trial.copy()
            best_out = out

        if (
            i + 1
        ) % max(
            1,
            n_random // 8
        ) == 0:

            log.info(
                f"[{variant_name}] "
                f"rand {i+1}/{n_random} "
                f"best={best_score:.5f}"
            )

    for j in range(
        n_refine
    ):

        improved = False

        for key in opt_keys:

            if key not in bounds:
                continue

            mode, lo, hi = (
                bounds[key]
            )

            center = float(
                best_params[key]
            )

            if mode == "log":

                candidates = [
                    center * 0.80,
                    center * 0.90,
                    center,
                    center * 1.10,
                    center * 1.25
                ]

            else:

                span = (
                    hi - lo
                )

                candidates = [
                    center
                    - 0.07
                    * span,

                    center
                    - 0.03
                    * span,

                    center,

                    center
                    + 0.03
                    * span,

                    center
                    + 0.07
                    * span
                ]

            for value in candidates:

                trial = dict(
                    best_params
                )

                trial[key] = float(
                    np.clip(
                        value,
                        lo,
                        hi
                    )
                )

                score, details, out = evaluate(
                    trial
                )

                if score < best_score:

                    best_score = score
                    best_params = trial.copy()
                    best_out = out
                    improved = True

        log.info(
            f"[{variant_name}] "
            f"refine {j+1}/{n_refine} "
            f"best={best_score:.5f}"
        )

        if not improved:
            break

    return (
        best_params,
        best_out,
        best_score
    )



# Target files to retrieve


def find_target_file(
    all_csvs,
    filename
):

    exact = [
        p
        for p in all_csvs
        if (
            os.path.basename(
                p
            ).lower()
            == filename.lower()
        )
    ]

    if exact:
        return exact[0]

    raise FileNotFoundError(
        f"No encontré {filename}"
    )


def resolve_targets(all_csvs):

    return (
        TARGET_SCHEME,

        find_target_file(
            all_csvs,
            f"Water_{TARGET_SCHEME}.csv"
        ),

        find_target_file(
            all_csvs,
            f"Ri_{TARGET_SCHEME}.csv"
        ),

        find_target_file(
            all_csvs,
            f"PBL_{TARGET_SCHEME}.csv"
        ),
    )






def build_variants():

    return {

        "conservative": {

            "source_gain": 0.010,

            "source_peak": 0.95,

            "source_pulse": 0.60,

            "reservoir_capacity": 1.5,

            "reservoir_decay": 0.88,

            "reservoir_input_frac": 0.30,

            "ql_to_liq_scale": 12.0,

            "liq_direct_scale": 2.0,

            "liq_event_gain": 2.5,

            "c_km": 0.07,

            "ell_0": 90.0,

            "pbl_forcing_gain": 18.0,

            "pbl_event_gain": 35.0,

            "pbl_ri_gain": 12.0,

            "pbl_omega": 0.18,

            "pbl_zeta": 0.45,

            "ri_scale": 10.0,

            "ri_pulse_gain": 1.8,

            "ri_relax_alpha": 0.16,

            "water_out_phys_weight": 0.84,

            "water_out_target_relax": 0.09,

            "water_out_slope_gain": 0.14,

            "pbl_out_phys_weight": 0.89,

            "pbl_out_target_relax": 0.055,

            "pbl_out_high_relax_gain": 0.12,

            "pbl_out_low_relax_gain": 0.09,

            "pbl_out_event_relax_gain": 0.08,

            "pbl_out_bias_gain": 0.18,

            "ri_out_phys_weight": 0.72,

            "ri_out_target_relax": 0.24,

            "ri_out_bias_gain": 0.25,
        },

        "balanced": {

            "source_gain": 0.014,

            "source_peak": 1.12,

            "source_pulse": 0.74,

            "reservoir_capacity": 2.4,

            "reservoir_decay": 0.87,

            "reservoir_input_frac": 0.42,

            "ql_to_liq_scale": 18.0,

            "liq_direct_scale": 5.0,

            "liq_event_gain": 5.0,

            "c_km": 0.11,

            "ell_0": 55.0,

            "pbl_forcing_gain": 34.0,

            "pbl_event_gain": 72.0,

            "pbl_ri_gain": 10.0,

            "pbl_omega": 0.30,

            "pbl_zeta": 0.45,

            "ri_scale": 12.0,

            "ri_pulse_gain": 3.5,

            "ri_relax_alpha": 0.12,

            "water_out_phys_weight": 0.76,

            "water_out_target_relax": 0.12,

            "water_out_slope_gain": 0.18,

            "pbl_out_phys_weight": 0.84,

            "pbl_out_target_relax": 0.075,

            "pbl_out_high_relax_gain": 0.15,

            "pbl_out_low_relax_gain": 0.12,

            "pbl_out_event_relax_gain": 0.11,

            "pbl_out_bias_gain": 0.24,

            "ri_out_phys_weight": 0.62,

            "ri_out_target_relax": 0.29,

            "ri_out_bias_gain": 0.31,
        },

        "aggressive": {

            "source_gain": 0.018,

            "source_peak": 1.35,

            "source_pulse": 0.90,

            "reservoir_capacity": 3.2,

            "reservoir_decay": 0.84,

            "reservoir_input_frac": 0.50,

            "ql_to_liq_scale": 22.0,

            "liq_direct_scale": 7.0,

            "liq_event_gain": 7.0,

            "c_km": 0.13,

            "ell_0": 45.0,

            "pbl_forcing_gain": 36.0,

            "pbl_event_gain": 70.0,

            "pbl_ri_gain": 18.0,

            "pbl_omega": 0.25,

            "pbl_zeta": 0.36,

            "ri_scale": 10.0,

            "ri_pulse_gain": 3.0,

            "ri_relax_alpha": 0.22,

            "water_out_phys_weight": 0.68,

            "water_out_target_relax": 0.17,

            "water_out_slope_gain": 0.23,

            "pbl_out_phys_weight": 0.78,

            "pbl_out_target_relax": 0.095,

            "pbl_out_high_relax_gain": 0.18,

            "pbl_out_low_relax_gain": 0.15,

            "pbl_out_event_relax_gain": 0.14,

            "pbl_out_bias_gain": 0.30,

            "ri_out_phys_weight": 0.55,

            "ri_out_target_relax": 0.34,

            "ri_out_bias_gain": 0.36,
        },
    }



# Bounds


def make_bounds():

    return {

        # water
        "source_gain":
            ("lin", 0.004, 0.040),

        "source_peak":
            ("lin", 0.50, 2.00),

        "source_pulse":
            ("lin", 0.20, 1.50),

        "reservoir_capacity":
            ("lin", 0.5, 4.0),

        "reservoir_decay":
            ("lin", 0.65, 0.98),

        "reservoir_input_frac":
            ("lin", 0.10, 0.95),

        "ql_to_liq_scale":
            ("lin", 6.0, 45.0),

        "liq_direct_scale":
            ("lin", 0.0, 15.0),

        "liq_event_gain":
            ("lin", 0.0, 15.0),

        # PBL 
        "c_km":
            ("lin", 0.03, 0.22),

        "ell_0":
            ("lin", 15.0, 150.0),

        "pbl_forcing_gain":
            ("lin", 4.0, 60.0),

        "pbl_event_gain":
            ("lin", 0.0, 120.0),

        "pbl_ri_gain":
            ("lin", 0.0, 60.0),

        "pbl_omega":
            ("lin", 0.05, 0.60),

        "pbl_zeta":
            ("lin", 0.05, 1.20),

        # Ri 
        "ri_scale":
            ("lin", 4.0, 20.0),

        "ri_pulse_gain":
            ("lin", 0.0, 8.0),

        "ri_relax_alpha":
            ("lin", 0.05, 0.55),

        
        "water_out_phys_weight":
            ("lin", 0.62, 0.92),

        "water_out_target_relax":
            ("lin", 0.05, 0.24),

        "water_out_event_relax":
            ("lin", 0.01, 0.13),

        "water_out_slope_gain":
            ("lin", 0.07, 0.30),

        "water_out_curvature_gain":
            ("lin", 0.05, 0.30),

        "water_out_memory_gain":
            ("lin", 0.10, 0.55),

       
        "pbl_out_phys_weight":
            ("lin", 0.70, 0.94),

        "pbl_out_error_phys_drop":
            ("lin", 0.05, 0.24),

        "pbl_out_event_phys_drop":
            ("lin", 0.02, 0.15),

        "pbl_out_target_relax":
            ("lin", 0.035, 0.15),

        "pbl_out_high_relax_gain":
            ("lin", 0.05, 0.28),

        "pbl_out_low_relax_gain":
            ("lin", 0.04, 0.24),

        "pbl_out_event_relax_gain":
            ("lin", 0.03, 0.22),

        "pbl_out_slope_gain":
            ("lin", 0.015, 0.13),

        "pbl_out_curvature_gain":
            ("lin", 0.015, 0.14),

        "pbl_out_memory_gain":
            ("lin", 0.08, 0.48),

        "pbl_out_bias_gain":
            ("lin", 0.08, 0.45),

        
        "ri_out_phys_weight":
            ("lin", 0.46, 0.78),

        "ri_out_high_phys_drop":
            ("lin", 0.00, 0.15),

        "ri_out_target_relax":
            ("lin", 0.18, 0.40),

        "ri_out_positive_relax_gain":
            ("lin", 0.00, 0.16),

        "ri_out_slope_gain":
            ("lin", 0.015, 0.10),

        "ri_out_memory_gain":
            ("lin", 0.35, 0.90),

        "ri_out_bias_gain":
            ("lin", 0.15, 0.48),
    }



# MAIN (starts program)


def main():

    folder = Path(
        FOLDER
    )

    if not folder.exists():

        raise FileNotFoundError(
            "Folder not found: "
            + FOLDER
        )

    all_csvs = sorted(
        str(p)
        for p in folder.glob(
            "*.csv"
        )
    )

    if not all_csvs:

        raise FileNotFoundError(
            "No CSVs in folder."
        )

    (
        scheme,
        water_file,
        ri_file,
        pbl_file
    ) = resolve_targets(
        all_csvs
    )

    log.info(
        f"Target scheme fixed: "
        f"{scheme}"
    )

    log.info(
        f"Targets: "
        f"water={os.path.basename(water_file)}, "
        f"ri={os.path.basename(ri_file)}, "
        f"pbl={os.path.basename(pbl_file)}"
    )

    cont_candidates = [
        p
        for p in all_csvs
        if (
            "cont_agua"
            in os.path.basename(
                p
            ).lower()
        )
    ]

    if not cont_candidates:

        raise FileNotFoundError(
            "Cont_agua.csv not found."
        )

    _, ql_init_series = (
        read_force_second_column_v3(
            cont_candidates[0],
            debug=False
        )
    )

    water_x, water_y = (
        read_force_second_column_v3(
            water_file,
            debug=False
        )
    )

    ri_x, ri_y = (
        read_force_second_column_v3(
            ri_file,
            debug=False
        )
    )

    pbl_x, pbl_y = (
        read_force_second_column_v3(
            pbl_file,
            debug=False
        )
    )

    N_target = min(
        len(water_y),
        len(ri_y),
        len(pbl_y)
    )

    target_water = np.asarray(
        water_y[
            :N_target
        ],
        dtype=float
    )

    target_ri = np.asarray(
        ri_y[
            :N_target
        ],
        dtype=float
    )

    target_pbl = np.asarray(
        pbl_y[
            :N_target
        ],
        dtype=float
    )

    water_x = resample_to_len(
        water_x,
        N_target
    )

    ri_x = resample_to_len(
        ri_x,
        N_target
    )

    pbl_x = resample_to_len(
        pbl_x,
        N_target
    )

    
    # Temperature per altitude files
    

    def is_temperature_file(path):

        name = os.path.basename(
            path
        ).lower()

        if re.search(
            r"^(water|ri|pbl)_",
            name
        ):
            return False

        if "cont_agua" in name:
            return False

        return (
            "_m_"
            in name
            and name.endswith(
                ".csv"
            )
        )

    temperature_files = [
        p
        for p in all_csvs
        if is_temperature_file(
            p
        )
    ]

    log.info(
        "Temp candidates: "
        + str(
            [
                os.path.basename(
                    p
                )
                for p
                in temperature_files
            ]
        )
    )

    variants = build_variants()

    bounds = make_bounds()

    WATER_PHYSICAL_KEYS = [

        "source_gain",

        "source_peak",

        "source_pulse",

        "reservoir_capacity",

        "reservoir_decay",

        "reservoir_input_frac",

        "ql_to_liq_scale",

        "liq_direct_scale",

        "liq_event_gain",
    ]

    PBL_PHYSICAL_KEYS = [

        "c_km",

        "ell_0",

        "pbl_forcing_gain",

        "pbl_event_gain",

        "pbl_ri_gain",

        "pbl_omega",

        "pbl_zeta",

        "ri_scale",

        "ri_pulse_gain",

        "ri_relax_alpha",
    ]

    OUTPUT_KEYS = [

        # Water
        "water_out_phys_weight",

        "water_out_target_relax",

        "water_out_event_relax",

        "water_out_slope_gain",

        "water_out_curvature_gain",

        "water_out_memory_gain",

        # PBL
        "pbl_out_phys_weight",

        "pbl_out_error_phys_drop",

        "pbl_out_event_phys_drop",

        "pbl_out_target_relax",

        "pbl_out_high_relax_gain",

        "pbl_out_low_relax_gain",

        "pbl_out_event_relax_gain",

        "pbl_out_slope_gain",

        "pbl_out_curvature_gain",

        "pbl_out_memory_gain",

        "pbl_out_bias_gain",

        # Ri
        "ri_out_phys_weight",

        "ri_out_high_phys_drop",

        "ri_out_target_relax",

        "ri_out_positive_relax_gain",

        "ri_out_slope_gain",

        "ri_out_memory_gain",

        "ri_out_bias_gain",
    ]

    for temp_file in temperature_files:

        basename = os.path.basename(
            temp_file
        )

        log.info(
            f"Processing: {basename}"
        )

        results = {}

        for (
            variant_name,
            base
        ) in variants.items():

            seed_offset = {

                "conservative":
                    0,

                "balanced":
                    1000,

                "aggressive":
                    2000,

            }[
                variant_name
            ]

            if DO_CALIBRATION:

                t0 = time.time()

                p1, o1, s1 = calibrate_stage(

                    temp_file,

                    variant_name,

                    ql_init_series,

                    target_water,

                    target_pbl,

                    target_ri,

                    base,

                    WATER_PHYSICAL_KEYS,

                    bounds,

                    N_RANDOM_STAGE1,

                    N_REFINE_STAGE1,

                    weights=(
                        0.68,
                        0.20,
                        0.12
                    ),

                    seed=(
                        RANDOM_SEED
                        + 11
                        + seed_offset
                    ),

                    interp_len=N_target,

                    guidance_strengths=(
                        GUIDANCE_STAGE_STRENGTHS[
                            "stage1"
                        ]
                    )
                )

                p2, o2, s2 = calibrate_stage(

                    temp_file,

                    variant_name,

                    ql_init_series,

                    target_water,

                    target_pbl,

                    target_ri,

                    p1,

                    PBL_PHYSICAL_KEYS,

                    bounds,

                    N_RANDOM_STAGE2,

                    N_REFINE_STAGE2,

                    weights=(
                        0.18,
                        0.62,
                        0.20
                    ),

                    seed=(
                        RANDOM_SEED
                        + 23
                        + seed_offset
                    ),

                    interp_len=N_target,

                    guidance_strengths=(
                        GUIDANCE_STAGE_STRENGTHS[
                            "stage2"
                        ]
                    )
                )

                
                p3, o3, s3 = calibrate_stage(

                    temp_file,

                    variant_name,

                    ql_init_series,

                    target_water,

                    target_pbl,

                    target_ri,

                    p2,

                    OUTPUT_KEYS,

                    bounds,

                    N_RANDOM_STAGE3,

                    N_REFINE_STAGE3,

                    weights=WEIGHTS_GLOBAL,

                    seed=(
                        RANDOM_SEED
                        + 37
                        + seed_offset
                    ),

                    interp_len=N_target,

                    guidance_strengths=(
                        GUIDANCE_STAGE_STRENGTHS[
                            "stage3"
                        ]
                    )
                )

                results[
                    variant_name
                ] = (
                    p3,
                    o3
                )

                log.info(
                    f"{basename} / "
                    f"{variant_name}: "
                    f"stage1={s1:.5f}, "
                    f"stage2={s2:.5f}, "
                    f"stage3={s3:.5f}, "
                    f"time="
                    f"{time.time()-t0:.1f}s"
                )

            else:

                out = run_case(

                    temp_file,

                    base,

                    ql_init_series,

                    variant_name=variant_name,

                    interp_to_len=N_target,

                    guidance_targets={

                        "liq":
                            target_water,

                        "pbl":
                            target_pbl,

                        "ri":
                            target_ri,
                    },

                    guidance_strengths=(
                        GUIDANCE_STAGE_STRENGTHS[
                            "stage3"
                        ]
                    )
                )

                results[
                    variant_name
                ] = (
                    base,
                    out
                )

        colors = {

            "conservative":
                "C0",

            "balanced":
                "C1",

            "aggressive":
                "C2",
        }

        
        # Water plot
        

        plt.figure(
            figsize=(
                12,
                5
            )
        )

        plt.plot(
            water_x,
            target_water,
            "k-",
            linewidth=2,
            label="Target water (raw)"
        )

        for (
            variant,
            (
                params_used,
                out
            )
        ) in results.items():

            if out is None:
                continue

            plt.plot(
                water_x,
                resample_to_len(
                    out["liq"],
                    N_target
                ),
                "--",
                color=colors[
                    variant
                ],
                label=variant
            )

        plt.title(
            f"Columna agua - {basename}"
        )

        plt.xlabel(
            "X (target)"
        )

        plt.ylabel(
            "Columna agua"
        )

        plt.grid(
            True,
            alpha=0.4
        )

        plt.legend()

        plt.tight_layout()

        plt.show()

        
        # Ri plot
        

        plt.figure(
            figsize=(
                12,
                5
            )
        )

        plt.plot(
            ri_x,
            target_ri,
            "k-",
            linewidth=2,
            label="Target Ri (raw)"
        )

        for (
            variant,
            (
                params_used,
                out
            )
        ) in results.items():

            if out is None:
                continue

            plt.plot(
                ri_x,
                resample_to_len(
                    out["ri"],
                    N_target
                ),
                "--",
                color=colors[
                    variant
                ],
                label=variant
            )

        plt.title(
            f"Richardson híbrido - "
            f"{basename}"
        )

        plt.xlabel(
            "X (target)"
        )

        plt.ylabel(
            "Ri"
        )

        plt.ylim(
            -35,
            35
        )

        plt.grid(
            True,
            alpha=0.4
        )

        plt.legend()

        plt.tight_layout()

        plt.show()

       
        # PBL plot
        

        plt.figure(
            figsize=(
                12,
                5
            )
        )

        plt.plot(
            pbl_x,
            target_pbl,
            "k-",
            linewidth=2,
            label="Target PBL (raw)"
        )

        for (
            variant,
            (
                params_used,
                out
            )
        ) in results.items():

            if out is None:
                continue

            plt.plot(
                pbl_x,
                resample_to_len(
                    out[
                        "pbl_comb"
                    ],
                    N_target
                ),
                "--",
                color=colors[
                    variant
                ],
                label=variant
            )

        plt.title(
            f"PBL - {basename}"
        )

        plt.xlabel(
            "X (target)"
        )

        plt.ylabel(
            "PBL [m]"
        )

        plt.ylim(
            0,
            max(
                220.0,
                1.08
                * np.max(
                    target_pbl
                )
            )
        )

        plt.grid(
            True,
            alpha=0.4
        )

        plt.legend()

        plt.tight_layout()

        plt.show()


        # ====================================================
        # DERIVATIVES, ERROR PERCENTAGES, AND BRUNT-VAISALA
        # ====================================================

        dwater_target_dx = derivative_wrt_x(
            target_water,
            water_x
        )

        dri_target_dx = derivative_wrt_x(
            target_ri,
            ri_x
        )

        dpbl_target_dx = derivative_wrt_x(
            target_pbl,
            pbl_x
        )

        
        d2water_target_dx2 = derivative_wrt_x(
            dwater_target_dx,
            water_x
        )

        if PRINT_ERROR_METRICS:

            log.info(
                f"===== Error/derivative summary: {basename} ====="
            )

            log.info(
                "Percentage errors: NMAE% and NRMSE% are normalized "
                "by target range; MAPE(valid)% excludes target values near zero."
            )

        analysis_by_variant = {}

        for (
            variant,
            (
                params_used,
                out
            )
        ) in results.items():

            if out is None:
                continue

            water_model = resample_to_len(
                out["liq"],
                N_target
            )

            ri_model = resample_to_len(
                out["ri"],
                N_target
            )

            pbl_model = resample_to_len(
                out["pbl_comb"],
                N_target
            )

            N_model = resample_to_len(
                out["N"],
                N_target
            )

            dwater_dx = derivative_wrt_x(
                water_model,
                water_x
            )

            dri_dx = derivative_wrt_x(
                ri_model,
                ri_x
            )

            dpbl_dx = derivative_wrt_x(
                pbl_model,
                pbl_x
            )

            d2water_dx2 = derivative_wrt_x(
                dwater_dx,
                water_x
            )

            dN_dx = derivative_wrt_x(
                N_model,
                water_x
            )

            water_error = percentage_error_metrics(
                water_model,
                target_water
            )

            ri_error = percentage_error_metrics(
                ri_model,
                target_ri
            )

            pbl_error = percentage_error_metrics(
                pbl_model,
                target_pbl
            )

            analysis_by_variant[variant] = {
                "water": water_model,
                "ri": ri_model,
                "pbl": pbl_model,
                "N": N_model,
                "dwater_dx": dwater_dx,
                "dri_dx": dri_dx,
                "dpbl_dx": dpbl_dx,
                "d2water_dx2": d2water_dx2,
                "dN_dx": dN_dx,
                "water_error": water_error,
                "ri_error": ri_error,
                "pbl_error": pbl_error,
            }

            if PRINT_ERROR_METRICS:

                log.info(
                    f"{basename} / {variant} / WATER: "
                    f"NMAE={water_error['NMAE_percent']:.2f}% | "
                    f"NRMSE={water_error['NRMSE_percent']:.2f}% | "
                    f"MAPE(valid)={water_error['MAPE_valid_percent']:.2f}% | "
                    f"bias={water_error['bias']:.3f} | "
                    f"final dW/dX={final_segment_slope(water_model, water_x):.4f}"
                )

                log.info(
                    f"{basename} / {variant} / RI: "
                    f"NMAE={ri_error['NMAE_percent']:.2f}% | "
                    f"NRMSE={ri_error['NRMSE_percent']:.2f}% | "
                    f"MAPE(valid)={ri_error['MAPE_valid_percent']:.2f}% | "
                    f"bias={ri_error['bias']:.3f} | "
                    f"final dRi/dX={final_segment_slope(ri_model, ri_x):.4f}"
                )

                log.info(
                    f"{basename} / {variant} / PBL: "
                    f"NMAE={pbl_error['NMAE_percent']:.2f}% | "
                    f"NRMSE={pbl_error['NRMSE_percent']:.2f}% | "
                    f"MAPE(valid)={pbl_error['MAPE_valid_percent']:.2f}% | "
                    f"bias={pbl_error['bias']:.3f} m | "
                    f"final dPBL/dX={final_segment_slope(pbl_model, pbl_x):.4f} m/X"
                )

                log.info(
                    f"{basename} / {variant} / BRUNT-VAISALA: "
                    f"mean N={safe_mean(N_model, 0.0):.6f} s^-1 | "
                    f"max N={safe_max(N_model, 0.0):.6f} s^-1 | "
                    f"final dN/dX={final_segment_slope(N_model, water_x):.6f}"
                )

        if PRINT_ERROR_METRICS:

            log.info(
                f"{basename} / TARGET final slopes: "
                f"dW/dX={final_segment_slope(target_water, water_x):.4f} | "
                f"dRi/dX={final_segment_slope(target_ri, ri_x):.4f} | "
                f"dPBL/dX={final_segment_slope(target_pbl, pbl_x):.4f} m/X"
            )

        if PLOT_RATE_DIAGNOSTICS:

            # dW/dX
            plt.figure(
                figsize=(
                    12,
                    5
                )
            )

            plt.plot(
                water_x,
                dwater_target_dx,
                "k-",
                linewidth=2,
                label="Target dW/dX"
            )

            for variant, data in analysis_by_variant.items():
                plt.plot(
                    water_x,
                    data["dwater_dx"],
                    "--",
                    color=colors[variant],
                    label=variant
                )

            plt.axhline(0.0, linewidth=1)
            plt.title(f"Tasa de cambio columna de agua dW/dX - {basename}")
            plt.xlabel("X (target)")
            plt.ylabel("dW/dX")
            plt.grid(True, alpha=0.4)
            plt.legend()
            plt.tight_layout()
            plt.show()

            # d2W/dX2
            plt.figure(
                figsize=(
                    12,
                    5
                )
            )

            plt.plot(
                water_x,
                d2water_target_dx2,
                "k-",
                linewidth=2,
                label="Target d²W/dX²"
            )

            for variant, data in analysis_by_variant.items():
                plt.plot(
                    water_x,
                    data["d2water_dx2"],
                    "--",
                    color=colors[variant],
                    label=variant
                )

            plt.axhline(0.0, linewidth=1)
            plt.title(f"Aceleración columna de agua d²W/dX² - {basename}")
            plt.xlabel("X (target)")
            plt.ylabel("d²W/dX²")
            plt.grid(True, alpha=0.4)
            plt.legend()
            plt.tight_layout()
            plt.show()

            # dRi/dX
            plt.figure(
                figsize=(
                    12,
                    5
                )
            )

            plt.plot(
                ri_x,
                dri_target_dx,
                "k-",
                linewidth=2,
                label="Target dRi/dX"
            )

            for variant, data in analysis_by_variant.items():
                plt.plot(
                    ri_x,
                    data["dri_dx"],
                    "--",
                    color=colors[variant],
                    label=variant
                )

            plt.axhline(0.0, linewidth=1)
            plt.title(f"Tasa de cambio Richardson dRi/dX - {basename}")
            plt.xlabel("X (target)")
            plt.ylabel("dRi/dX")
            plt.grid(True, alpha=0.4)
            plt.legend()
            plt.tight_layout()
            plt.show()

            
            plt.figure(
                figsize=(
                    12,
                    5
                )
            )

            plt.plot(
                pbl_x,
                dpbl_target_dx,
                "k-",
                linewidth=2,
                label="Target dPBL/dX"
            )

            for variant, data in analysis_by_variant.items():
                plt.plot(
                    pbl_x,
                    data["dpbl_dx"],
                    "--",
                    color=colors[variant],
                    label=variant
                )

            plt.axhline(0.0, linewidth=1)
            plt.title(f"Tasa de cambio PBL dPBL/dX - {basename}")
            plt.xlabel("X (target)")
            plt.ylabel("dPBL/dX [m/X]")
            plt.grid(True, alpha=0.4)
            plt.legend()
            plt.tight_layout()
            plt.show()

        if PLOT_BRUNT_VAISALA:

            
            plt.figure(
                figsize=(
                    12,
                    5
                )
            )

            for variant, data in analysis_by_variant.items():
                plt.plot(
                    water_x,
                    data["N"],
                    "--",
                    color=colors[variant],
                    label=variant
                )

            plt.title(f"Frecuencia de Brunt-Väisälä - {basename}")
            plt.xlabel("X (target)")
            plt.ylabel("N [s⁻¹]")
            plt.grid(True, alpha=0.4)
            plt.legend()
            plt.tight_layout()
            plt.show()

           
            plt.figure(
                figsize=(
                    12,
                    5
                )
            )

            for variant, data in analysis_by_variant.items():
                plt.plot(
                    water_x,
                    data["dN_dx"],
                    "--",
                    color=colors[variant],
                    label=variant
                )

            plt.axhline(0.0, linewidth=1)
            plt.title(f"Tasa de cambio Brunt-Väisälä dN/dX - {basename}")
            plt.xlabel("X (target)")
            plt.ylabel("dN/dX [s⁻¹/X]")
            plt.grid(True, alpha=0.4)
            plt.legend()
            plt.tight_layout()
            plt.show()


        if PLOT_PHYSICAL_DIAGNOSTICS:

            diagnostics_to_plot = [

                (
                    "Water physical",
                    "liq_physical",
                    water_x,
                    target_water
                ),

                (
                    "PBL physical",
                    "pbl_physical",
                    pbl_x,
                    target_pbl
                ),

                (
                    "Ri physical",
                    "ri_physical",
                    ri_x,
                    target_ri
                ),
            ]

            for (
                title,
                key,
                xvals,
                target
            ) in diagnostics_to_plot:

                plt.figure(
                    figsize=(
                        12,
                        5
                    )
                )

                plt.plot(
                    xvals,
                    target,
                    "k-",
                    linewidth=2,
                    label="Target"
                )

                for (
                    variant,
                    (
                        params_used,
                        out
                    )
                ) in results.items():

                    if out is None:
                        continue

                    plt.plot(
                        xvals,
                        resample_to_len(
                            out[key],
                            N_target
                        ),
                        "--",
                        color=colors[
                            variant
                        ],
                        label=variant
                    )

                plt.title(
                    f"{title} - {basename}"
                )

                plt.grid(
                    True
                )

                plt.legend()

                plt.tight_layout()

                plt.show()

    log.info(
        "All done."
    )




if __name__ == "__main__":
    main()
```
